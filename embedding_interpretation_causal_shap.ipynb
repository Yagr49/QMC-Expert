# Интерпретация эмбеддингов: MoLFormer vs COMA

Сравнение **MoLFormer (768d)** и **латентного пространства COMA (256d)** после контрастивного переноса:

1. **SHAP** — ассоциативная важность PCA-компонент в CatBoost
2. **Causal Analysis** — `LinearDML` (econml): ATE каждой компоненты при контроле остальных
3. **Выравнивание пространств** — CCA, CKA, доля «шума» MoLFormer, не покрываемого COMA

Таргеты: SA_MIC, SA_MBC, EC_MIC, EC_MBC (log2).

Скрипт: `scripts/embedding_causal_shap_analysis.py`

```python
import subprocess, sys
from pathlib import Path

BASE = Path('/Users/egorilin/Desktop/COMA')
script = BASE / 'scripts' / 'embedding_causal_shap_analysis.py'
result = subprocess.run([sys.executable, str(script)], capture_output=True, text=True)
print(result.stdout)
if result.returncode != 0:
    print('STDERR:', result.stderr)
    raise RuntimeError(f'Script failed with code {result.returncode}')
```

```python
from pathlib import Path
from IPython.display import Image, display, Markdown
import pandas as pd

OUT = Path('/Users/egorilin/Desktop/COMA/outputs_embedding_interpretation')

display(Markdown('## Отчёт'))
print((OUT / 'REPORT.txt').read_text(encoding='utf-8'))

display(Markdown('## Сводка по таргетам'))
display(pd.read_excel(OUT / 'summary_per_target.xlsx'))

display(Markdown('## CV R²'))
display(Image(filename=str(OUT / 'cv_r2_comparison.png')))
```

```python
from pathlib import Path
from IPython.display import Image, display, Markdown

OUT = Path('/Users/egorilin/Desktop/COMA/outputs_embedding_interpretation')
targets = ['SA_MIC', 'SA_MBC', 'EC_MIC', 'EC_MBC']

for sn in targets:
    shap_p = OUT / f'shap_compare_{sn}.png'
    causal_p = OUT / f'causal_compare_{sn}.png'
    if shap_p.exists():
        display(Markdown(f'### {sn}: SHAP vs Causal'))
        display(Image(filename=str(shap_p), width=900))
        if causal_p.exists():
            display(Image(filename=str(causal_p), width=900))
```
