```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error, r2_score
from sklearn.preprocessing import StandardScaler
from catboost import CatBoostRegressor
from sklearn.neural_network import MLPRegressor
import warnings
warnings.filterwarnings('ignore')

# ═══════════════════════════════════════════════════════════════
# 0. ДАННЫЕ
# ═══════════════════════════════════════════════════════════════

df_old = pd.read_excel('/Users/egorilin/Desktop/COMA/Data_molformer_conc.xlsx')

feature_cols = [
    'Molecular weight', 'LogP', 'mmff94_dipole', 'gasteiger_dipole', 'eem2015bm_dipole',
    'Molformer_157', 'Molformer_197', 'Molformer_668', 'Molformer_85',
    'Molformer_127', 'Molformer_732', 'Molformer_543', 'Molformer_729',
    'Molformer_337', 'Molformer_686'
]

target_columns = [
    'S. aureus ATCC 43300 Activity MIC, мг/л',
    'S. aureus ATCC 43300 Activity MBC, мг/л',
    'E. coli ATCC 25922 Activity MIC, мг/л',
    'E. coli ATCC 25922 Activity MBC, мг/л'
]
short_names = ['SA_MIC', 'SA_MBC', 'EC_MIC', 'EC_MBC']

X_all = df_old[feature_cols].values

# ═══════════════════════════════════════════════════════════════
# 1. ПОИСК ЛУЧШЕГО random_state ДЛЯ КАЖДОГО ТАРГЕТА
# ═══════════════════════════════════════════════════════════════

print("=" * 80)
print("ПОИСК ЛУЧШЕЙ РАЗБИВКИ (random_state 0..99)")
print("=" * 80)

SEEDS = range(100)
TEST_SIZE = 0.2

best_seeds = {}  # {target: best_seed}
all_search_results = []

for col, sn in zip(target_columns, short_names):
    valid_mask = ~df_old[col].isna()
    X = X_all[valid_mask]
    y = np.log2(df_old[col][valid_mask].values)

    print(f"\n{sn} (N={len(y)}):")

    best_r2 = -999
    best_seed = 0
    seed_results = []

    for seed in SEEDS:
        Xtr, Xte, ytr, yte = train_test_split(X, y, test_size=TEST_SIZE, random_state=seed)

        scaler = StandardScaler()
        Xtr_s = scaler.fit_transform(Xtr)
        Xte_s = scaler.transform(Xte)

        # CatBoost
        cb = CatBoostRegressor(iterations=500, learning_rate=0.05, depth=6,
                               verbose=False, random_seed=42)
        cb.fit(Xtr_s, ytr)
        yp_cb = cb.predict(Xte_s)
        r2_cb = r2_score(yte, yp_cb)
        rmse_cb = np.sqrt(mean_squared_error(yte, yp_cb))

        # MLP
        mlp = MLPRegressor(hidden_layer_sizes=(256, 128, 64), max_iter=1000,
                           early_stopping=True, validation_fraction=0.15,
                           learning_rate='adaptive', learning_rate_init=0.001,
                           random_state=42)
        mlp.fit(Xtr_s, ytr)
        yp_mlp = mlp.predict(Xte_s)
        r2_mlp = r2_score(yte, yp_mlp)
        rmse_mlp = np.sqrt(mean_squared_error(yte, yp_mlp))

        avg_r2 = (r2_cb + r2_mlp) / 2

        seed_results.append({
            'target': sn, 'seed': seed,
            'R2_cb': r2_cb, 'RMSE_cb': rmse_cb,
            'R2_mlp': r2_mlp, 'RMSE_mlp': rmse_mlp,
            'R2_avg': avg_r2
        })

        if avg_r2 > best_r2:
            best_r2 = avg_r2
            best_seed = seed

    best_seeds[sn] = best_seed
    all_search_results.extend(seed_results)

    # Топ-5 сидов
    sorted_results = sorted(seed_results, key=lambda x: x['R2_avg'], reverse=True)
    print(f"  Top-5 seeds (by avg R²):")
    for r in sorted_results[:5]:
        print(f"    seed={r['seed']:3d}  CB: R²={r['R2_cb']:.4f} RMSE={r['RMSE_cb']:.4f}  |  MLP: R²={r['R2_mlp']:.4f} RMSE={r['RMSE_mlp']:.4f}  |  avg={r['R2_avg']:.4f}")
    print(f"  >>> ЛУЧШИЙ seed={best_seed}, avg R²={best_r2:.4f}")

search_df = pd.DataFrame(all_search_results)

# ═══════════════════════════════════════════════════════════════
# 2. ЛУЧШИЙ ЕДИНЫЙ SEED (по среднему R² по всем таргетам)
# ═══════════════════════════════════════════════════════════════

print("\n" + "=" * 80)
print("ЛУЧШИЙ ЕДИНЫЙ SEED (средний R² по всем таргетам)")
print("=" * 80)

seed_avg = search_df.groupby('seed')['R2_avg'].mean().reset_index()
seed_avg = seed_avg.sort_values('R2_avg', ascending=False)

print("\nТоп-10 единых seeds:")
for _, row in seed_avg.head(10).iterrows():
    s = int(row['seed'])
    avg = row['R2_avg']
    details = search_df[search_df['seed'] == s]
    det_str = "  ".join([f"{r['target']}: CB={r['R2_cb']:.3f} MLP={r['R2_mlp']:.3f}"
                         for _, r in details.iterrows()])
    print(f"  seed={s:3d}  avg_R²={avg:.4f}  | {det_str}")

BEST_GLOBAL_SEED = int(seed_avg.iloc[0]['seed'])
print(f"\n>>> ЛУЧШИЙ ЕДИНЫЙ SEED = {BEST_GLOBAL_SEED}")

# ═══════════════════════════════════════════════════════════════
# 3. BASELINE с лучшим seed'ом
# ═══════════════════════════════════════════════════════════════

print("\n" + "=" * 80)
print(f"BASELINE: ЛУЧШИЙ SEED = {BEST_GLOBAL_SEED}")
print("=" * 80)

baseline_results = []

for col, sn in zip(target_columns, short_names):
    valid_mask = ~df_old[col].isna()
    X = X_all[valid_mask]
    y = np.log2(df_old[col][valid_mask].values)

    Xtr, Xte, ytr, yte = train_test_split(X, y, test_size=TEST_SIZE, random_state=BEST_GLOBAL_SEED)

    scaler = StandardScaler()
    Xtr_s = scaler.fit_transform(Xtr)
    Xte_s = scaler.transform(Xte)

    # CatBoost
    cb = CatBoostRegressor(iterations=500, learning_rate=0.05, depth=6,
                           verbose=False, random_seed=42)
    cb.fit(Xtr_s, ytr)
    yp_cb = cb.predict(Xte_s)

    # MLP
    mlp = MLPRegressor(hidden_layer_sizes=(256, 128, 64), max_iter=1000,
                       early_stopping=True, validation_fraction=0.15,
                       learning_rate='adaptive', learning_rate_init=0.001,
                       random_state=42)
    mlp.fit(Xtr_s, ytr)
    yp_mlp = mlp.predict(Xte_s)

    for model_name, yp in [('catboost', yp_cb), ('mlp', yp_mlp)]:
        r2 = r2_score(yte, yp)
        rmse = np.sqrt(mean_squared_error(yte, yp))
        baseline_results.append({
            'target': sn, 'model': model_name,
            'R2': r2, 'RMSE': rmse
        })
        print(f"  {sn:8s} | {model_name:10s} | R²={r2:.4f}  RMSE={rmse:.4f}")

baseline_df = pd.DataFrame(baseline_results)

# ═══════════════════════════════════════════════════════════════
# 4. MULTI-TASK с лучшим seed'ом
# ═══════════════════════════════════════════════════════════════

print("\n" + "=" * 80)
print(f"MULTI-TASK: ЛУЧШИЙ SEED = {BEST_GLOBAL_SEED}")
print("=" * 80)

n_tasks = len(target_columns)
n_feat = len(feature_cols)

rows_X, rows_y, rows_task, rows_orig_idx = [], [], [], []

for task_idx, col in enumerate(target_columns):
    valid_mask = ~df_old[col].isna()
    X = X_all[valid_mask]
    y = np.log2(df_old[col][valid_mask].values)
    indices = np.where(valid_mask)[0]

    onehot = np.zeros(n_tasks)
    onehot[task_idx] = 1.0

    for i in range(len(X)):
        rows_X.append(np.concatenate([X[i], onehot]))
        rows_y.append(y[i])
        rows_task.append(task_idx)
        rows_orig_idx.append(indices[i])

X_mt = np.array(rows_X)
y_mt = np.array(rows_y)
task_ids = np.array(rows_task)
orig_idx = np.array(rows_orig_idx)

# Сплит по молекулам с тем же seed'ом
unique_molecules = np.unique(orig_idx)
train_mols, test_mols = train_test_split(unique_molecules, test_size=TEST_SIZE,
                                          random_state=BEST_GLOBAL_SEED)
train_mols_set = set(train_mols)
test_mols_set = set(test_mols)

train_mask = np.array([oi in train_mols_set for oi in orig_idx])
test_mask  = np.array([oi in test_mols_set for oi in orig_idx])

Xtr, Xte = X_mt[train_mask], X_mt[test_mask]
ytr, yte = y_mt[train_mask], y_mt[test_mask]
tasks_te = task_ids[test_mask]

print(f"Train: {Xtr.shape[0]}, Test: {Xte.shape[0]}")

# Нормализация (только признаки, не OneHot)
scaler = StandardScaler()
Xtr_scaled = Xtr.copy()
Xte_scaled = Xte.copy()
Xtr_scaled[:, :n_feat] = scaler.fit_transform(Xtr[:, :n_feat])
Xte_scaled[:, :n_feat] = scaler.transform(Xte[:, :n_feat])

# CatBoost
cb = CatBoostRegressor(iterations=500, learning_rate=0.05, depth=6,
                       verbose=False, random_seed=42)
cb.fit(Xtr_scaled, ytr)
yp_cb = cb.predict(Xte_scaled)

# MLP
mlp = MLPRegressor(hidden_layer_sizes=(256, 128, 64), max_iter=1000,
                   early_stopping=True, validation_fraction=0.15,
                   learning_rate='adaptive', learning_rate_init=0.001,
                   random_state=42)
mlp.fit(Xtr_scaled, ytr)
yp_mlp = mlp.predict(Xte_scaled)

mt_results = []
for i, sn in enumerate(short_names):
    mask = tasks_te == i
    if mask.sum() < 2:
        continue
    yt_i = yte[mask]
    for model_name, yp in [('catboost', yp_cb), ('mlp', yp_mlp)]:
        yp_i = yp[mask]
        r2 = r2_score(yt_i, yp_i)
        rmse = np.sqrt(mean_squared_error(yt_i, yp_i))
        mt_results.append({'target': sn, 'model': model_name, 'R2': r2, 'RMSE': rmse})
        print(f"  {sn:8s} | {model_name:10s} | R²={r2:.4f}  RMSE={rmse:.4f}")

mt_df = pd.DataFrame(mt_results)

# ═══════════════════════════════════════════════════════════════
# 5. ИТОГОВОЕ СРАВНЕНИЕ
# ═══════════════════════════════════════════════════════════════

print("\n" + "=" * 80)
print(f"ИТОГ: BASELINE vs MULTI-TASK (seed={BEST_GLOBAL_SEED})")
print("=" * 80)

for model_name in ['catboost', 'mlp']:
    print(f"\n{'─'*70}")
    print(f"  {model_name.upper()}")
    print(f"{'─'*70}")
    print(f"  {'Target':8s} | {'BL_R²':>8s} | {'MT_R²':>8s} | {'ΔR²':>8s} | {'BL_RMSE':>9s} | {'MT_RMSE':>9s} | {'ΔRMSE':>8s}")
    print(f"  {'-'*8}-+-{'-'*8}-+-{'-'*8}-+-{'-'*8}-+-{'-'*9}-+-{'-'*9}-+-{'-'*8}")

    for sn in short_names:
        bl = baseline_df[(baseline_df['target'] == sn) & (baseline_df['model'] == model_name)]
        mt = mt_df[(mt_df['target'] == sn) & (mt_df['model'] == model_name)]
        if bl.empty or mt.empty:
            continue
        bl_r2   = bl['R2'].values[0]
        mt_r2   = mt['R2'].values[0]
        bl_rmse = bl['RMSE'].values[0]
        mt_rmse = mt['RMSE'].values[0]
        print(f"  {sn:8s} | {bl_r2:8.4f} | {mt_r2:8.4f} | {mt_r2-bl_r2:+8.4f} | {bl_rmse:9.4f} | {mt_rmse:9.4f} | {bl_rmse-mt_rmse:+8.4f}")
```

```python
# %% [markdown]
# # MULTI-TASK LEARNING: ПРЕДСКАЗАНИЕ АКТИВНОСТИ С ЭМБЕДДИНГАМИ
#
# ## Идея эксперимента:
#
# 1. **Baseline Regression**: Регрессия на каждый таргет отдельно
# 2. **Baseline Classification**: Классификация (дискретные таргеты) на каждый таргет
# 3. **Multi-Task Regression**: Объединённый датасет, метка класса как categorical фича
# 4. **Multi-Task Classification**: Объединённый датасет, классификация
# 5. **Advanced**: Метка класса через OpenAI embeddings (регрессия + классификация)
#
# ## Таргеты (4 штуки):
# - S. aureus ATCC 43300 Activity MIC
# - S. aureus ATCC 43300 Activity MBC
# - E. coli ATCC 25922 Activity MIC
# - E. coli ATCC 25922 Activity MBC

# %% [markdown]
# ## 📦 Импорты

# %%
import pandas as pd
import numpy as np
import torch
import torch.nn as nn
from torch.utils.data import Dataset, DataLoader
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.metrics import (
    mean_absolute_error, mean_squared_error, r2_score,
    accuracy_score, f1_score, classification_report, confusion_matrix
)
from catboost import CatBoostRegressor, CatBoostClassifier, Pool
from sklearn.linear_model import Ridge
from sklearn.ensemble import RandomForestRegressor
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import pearsonr
import pickle
import warnings
warnings.filterwarnings('ignore')

# OpenRouter для эмбеддингов меток (OpenAI-совместимый API)
from openai import OpenAI
import os

device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print(f"Device: {device}")

# %% [markdown]
# ## 🔑 Настройка OpenRouter API

# %%
OPENROUTER_API_KEY = "[REDACTED_API_KEY]"

# OpenRouter — OpenAI-совместимый API, просто другой base_url
# Прокси не нужен!
client = OpenAI(
    api_key=OPENROUTER_API_KEY,
    base_url="https://openrouter.ai/api/v1"
)

# Модель эмбеддингов через OpenRouter
EMBEDDING_MODEL = "openai/text-embedding-3-small"

def get_embedding(text, model=EMBEDDING_MODEL):
    """Получает эмбеддинг через OpenRouter"""
    text = text.replace("\n", " ")
    try:
        response = client.embeddings.create(input=[text], model=model)
        return response.data[0].embedding
    except Exception as e:
        print(f"❌ Ошибка OpenRouter API: {e}")
        raise

# Тест соединения
try:
    test_emb = get_embedding("test")
    print(f"✓ OpenRouter API работает")
    print(f"  Модель: {EMBEDDING_MODEL}")
    print(f"  Размерность эмбеддинга: {len(test_emb)}")
except Exception as e:
    print(f"❌ Не удалось подключиться к OpenRouter API")
    print(f"   Ошибка: {e}")

# %% [markdown]
# ## 📊 Загрузка данных

# %%
BASE_PATH = "/Users/egorilin/Desktop/COMA"

data_drop = pd.read_excel(f"{BASE_PATH}/Test_molformer_original_biocides.xlsx").drop(['Unnamed: 0'], axis=1)
target = pd.read_excel(f"{BASE_PATH}/Data_biocides.xlsx")

print(f"Данные: {data_drop.shape[0]} образцов, {data_drop.shape[1]} признаков")

target_columns = [
    'S. aureus ATCC 43300 Activity MIC, мг/л',
    'S. aureus ATCC 43300 Activity MBC, мг/л',
    'E. coli ATCC 25922 Activity MIC, мг/л',
    'E. coli ATCC 25922 Activity MBC, мг/л'
]

print(f"\nТаргеты:")
for col in target_columns:
    vals = sorted(target[col].dropna().unique())
    print(f"\n  {col}:")
    print(f"    Уникальные значения ({len(vals)}): {vals}")
    print(f"    NaN: {target[col].isna().sum()}")

# %% [markdown]
# ## 📋 Распределения таргетов

# %%
fig, axes = plt.subplots(2, 2, figsize=(15, 10))
axes = axes.flatten()

for idx, col in enumerate(target_columns):
    ax = axes[idx]
    values = target[col].dropna()
    ax.hist(values, bins=30, alpha=0.7, edgecolor='black')
    ax.set_xlabel('Значение (мг/л)', fontsize=11, fontweight='bold')
    ax.set_ylabel('Частота', fontsize=11, fontweight='bold')
    ax.set_title(col, fontsize=12, fontweight='bold')
    ax.grid(alpha=0.3)

plt.tight_layout()
plt.savefig(f'{BASE_PATH}/target_distributions.png', dpi=300, bbox_inches='tight')
plt.show()
print("✓ Сохранено: target_distributions.png")

# %% [markdown]
# ## 🔧 Загрузка предобученного энкодера

# %%
class AdvancedContrastiveEncoder(nn.Module):
    def __init__(self, input_dim, hidden_dims=[512, 384, 128], projection_dim=128, dropout=0.3):
        super(AdvancedContrastiveEncoder, self).__init__()

        layers = []
        prev_dim = input_dim
        for hidden_dim in hidden_dims:
            layers.extend([
                nn.Linear(prev_dim, hidden_dim),
                nn.BatchNorm1d(hidden_dim),
                nn.ReLU(),
                nn.Dropout(dropout)
            ])
            prev_dim = hidden_dim

        self.encoder = nn.Sequential(*layers)

        self.projection_head = nn.Sequential(
            nn.Linear(hidden_dims[-1], projection_dim),
            nn.BatchNorm1d(projection_dim),
            nn.ReLU(),
            nn.Dropout(dropout * 0.5),
            nn.Linear(projection_dim, projection_dim),
            nn.BatchNorm1d(projection_dim)
        )

    def forward(self, x, return_projection=True):
        features = self.encoder(x)
        if return_projection:
            projections = self.projection_head(features)
            return features, projections
        return features


encoder = AdvancedContrastiveEncoder(
    input_dim=768,
    hidden_dims=[512, 384, 128],
    projection_dim=128,
    dropout=0.3
).to(device)

encoder.load_state_dict(torch.load(f'{BASE_PATH}/encoder_borderline_smote.pth', map_location=device))
encoder.eval()

with open(f'{BASE_PATH}/scaler_classification.pkl', 'rb') as f:
    scaler = pickle.load(f)

print("✓ Энкодер и scaler загружены")

# %% [markdown]
# ## 🎯 Извлечение молекулярных эмбеддингов

# %%
def extract_embeddings(encoder, X_scaled):
    encoder.eval()
    X_tensor = torch.FloatTensor(X_scaled).to(device)
    with torch.no_grad():
        embeddings = encoder(X_tensor, return_projection=False)
        return embeddings.cpu().numpy()

data_scaled = scaler.transform(data_drop)
molecular_embeddings = extract_embeddings(encoder, data_scaled)
print(f"✓ Молекулярные эмбеддинги: {molecular_embeddings.shape}")

# %% [markdown]
# ---
# # 🧪 ЭКСПЕРИМЕНТ 1: BASELINE REGRESSION — Каждый таргет отдельно
# ---

# %%
print("=" * 80)
print("ЭКСПЕРИМЕНТ 1: BASELINE REGRESSION — КАЖДЫЙ ТАРГЕТ ОТДЕЛЬНО")
print("=" * 80)

baseline_reg_results = []

for target_col in target_columns:
    print(f"\n{'─'*60}")
    print(f"Таргет: {target_col}")

    valid_idx = ~target[target_col].isna()
    X_valid = molecular_embeddings[valid_idx]
    y_valid = np.log2(target[target_col][valid_idx].values)

    print(f"  Образцов: {len(y_valid)}, log2 диапазон: [{y_valid.min():.2f}, {y_valid.max():.2f}]")

    X_train, X_test, y_train, y_test = train_test_split(
        X_valid, y_valid, test_size=0.2, random_state=42
    )

    cb = CatBoostRegressor(iterations=500, learning_rate=0.05, depth=6,
                           verbose=False, random_seed=42)
    cb.fit(X_train, y_train)
    y_pred = cb.predict(X_test)

    mae = mean_absolute_error(y_test, y_pred)
    rmse = np.sqrt(mean_squared_error(y_test, y_pred))
    r2 = r2_score(y_test, y_pred)
    pr, _ = pearsonr(y_test, y_pred)

    short_name = target_col.split(' Activity ')[1].replace(', мг/л', '')
    organism = target_col.split(' Activity ')[0]

    print(f"  CatBoost: MAE={mae:.4f}  RMSE={rmse:.4f}  R²={r2:.4f}  Pearson={pr:.4f}")

    baseline_reg_results.append({
        'Target': short_name, 'Organism': organism,
        'N_samples': len(y_valid),
        'MAE': mae, 'RMSE': rmse, 'R²': r2, 'Pearson_r': pr
    })

baseline_reg_df = pd.DataFrame(baseline_reg_results)
print("\n\nBASELINE REGRESSION SUMMARY:")
print(baseline_reg_df.to_string(index=False))

# %% [markdown]
# ---
# # 🧪 ЭКСПЕРИМЕНТ 2: BASELINE CLASSIFICATION — Каждый таргет отдельно
# ---

# %%
print("\n" + "=" * 80)
print("ЭКСПЕРИМЕНТ 2: BASELINE CLASSIFICATION — КАЖДЫЙ ТАРГЕТ ОТДЕЛЬНО")
print("=" * 80)
print("(Таргеты дискретные → можно предсказывать класс напрямую)")

baseline_cls_results = []

for target_col in target_columns:
    print(f"\n{'─'*60}")
    print(f"Таргет: {target_col}")

    valid_idx = ~target[target_col].isna()
    X_valid = molecular_embeddings[valid_idx]
    y_valid_raw = target[target_col][valid_idx].values

    # Кодируем дискретные значения в классы
    le = LabelEncoder()
    y_valid_cls = le.fit_transform(y_valid_raw)
    n_classes = len(le.classes_)

    print(f"  Образцов: {len(y_valid_cls)}, классов: {n_classes}")
    print(f"  Классы (мг/л): {le.classes_}")

    # stratify если все классы имеют >= 2 образца, иначе без
    class_counts = np.bincount(y_valid_cls)
    can_stratify = np.min(class_counts[class_counts > 0]) >= 2

    X_train, X_test, y_train, y_test = train_test_split(
        X_valid, y_valid_cls, test_size=0.2, random_state=42,
        stratify=y_valid_cls if can_stratify else None
    )
    if not can_stratify:
        print(f"  ⚠️ Есть классы с 1 образцом — stratify отключён")

    cb_cls = CatBoostClassifier(
        iterations=500, learning_rate=0.05, depth=6,
        verbose=False, random_seed=42, auto_class_weights='Balanced'
    )
    cb_cls.fit(X_train, y_train)
    y_pred_cls = cb_cls.predict(X_test).flatten().astype(int)

    acc = accuracy_score(y_test, y_pred_cls)
    f1_macro = f1_score(y_test, y_pred_cls, average='macro', zero_division=0)
    f1_weighted = f1_score(y_test, y_pred_cls, average='weighted', zero_division=0)

    short_name = target_col.split(' Activity ')[1].replace(', мг/л', '')
    organism = target_col.split(' Activity ')[0]

    print(f"  CatBoost: Acc={acc:.4f}  F1_macro={f1_macro:.4f}  F1_weighted={f1_weighted:.4f}")

    baseline_cls_results.append({
        'Target': short_name, 'Organism': organism,
        'N_samples': len(y_valid_cls), 'N_classes': n_classes,
        'Accuracy': acc, 'F1_macro': f1_macro, 'F1_weighted': f1_weighted
    })

baseline_cls_df = pd.DataFrame(baseline_cls_results)
print("\n\nBASELINE CLASSIFICATION SUMMARY:")
print(baseline_cls_df.to_string(index=False))

# %% [markdown]
# ---
# # 🧪 ЭКСПЕРИМЕНТ 3: MULTI-TASK REGRESSION с Categorical метками
# ---

# %%
print("\n" + "=" * 80)
print("ЭКСПЕРИМЕНТ 3: MULTI-TASK REGRESSION С CATEGORICAL МЕТКАМИ")
print("=" * 80)

# Строим объединённый датасет
multi_task_data = []

for target_col in target_columns:
    valid_idx = ~target[target_col].isna()
    X_emb = molecular_embeddings[valid_idx]
    y_vals = np.log2(target[target_col][valid_idx].values)

    organism = target_col.split(' Activity ')[0]
    metric_type = target_col.split(' Activity ')[1].replace(', мг/л', '')

    for i in range(len(X_emb)):
        multi_task_data.append({
            'molecular_emb': X_emb[i],
            'organism': organism,
            'metric_type': metric_type,
            'target_value': y_vals[i],
            'target_raw': 2 ** y_vals[i]  # исходное значение для классификации
        })

print(f"✓ Multi-task датасет: {len(multi_task_data)} образцов")

X_molecular = np.array([d['molecular_emb'] for d in multi_task_data])
y_values = np.array([d['target_value'] for d in multi_task_data])

organisms = [d['organism'] for d in multi_task_data]
metric_types = [d['metric_type'] for d in multi_task_data]

organism_encoder = LabelEncoder()
metric_encoder = LabelEncoder()
organism_encoded = organism_encoder.fit_transform(organisms)
metric_encoded = metric_encoder.fit_transform(metric_types)

print(f"Организмы: {organism_encoder.classes_}")
print(f"Метрики: {metric_encoder.classes_}")

# DataFrame с categorical колонками
X_multi_cat_df = pd.DataFrame(X_molecular, columns=[f'emb_{i}' for i in range(128)])
X_multi_cat_df['organism'] = [str(x) for x in organism_encoded]
X_multi_cat_df['metric_type'] = [str(x) for x in metric_encoded]

print(f"Размерность: {X_multi_cat_df.shape}")

X_train_cat, X_test_cat, y_train_cat, y_test_cat = train_test_split(
    X_multi_cat_df, y_values, test_size=0.2, random_state=42
)

train_pool = Pool(data=X_train_cat, label=y_train_cat, cat_features=['organism', 'metric_type'])

cb_multi_cat = CatBoostRegressor(iterations=500, learning_rate=0.05, depth=6,
                                  verbose=False, random_seed=42)
cb_multi_cat.fit(train_pool)
y_pred_cat = cb_multi_cat.predict(X_test_cat)

mae_cat = mean_absolute_error(y_test_cat, y_pred_cat)
rmse_cat = np.sqrt(mean_squared_error(y_test_cat, y_pred_cat))
r2_cat = r2_score(y_test_cat, y_pred_cat)
pearson_cat, _ = pearsonr(y_test_cat, y_pred_cat)

print(f"\nMulti-Task Regression (Categorical):")
print(f"  MAE={mae_cat:.4f}  RMSE={rmse_cat:.4f}  R²={r2_cat:.4f}  Pearson={pearson_cat:.4f}")

# %% [markdown]
# ---
# # 🧪 ЭКСПЕРИМЕНТ 4: MULTI-TASK CLASSIFICATION с Categorical метками
# ---

# %%
print("\n" + "=" * 80)
print("ЭКСПЕРИМЕНТ 4: MULTI-TASK CLASSIFICATION С CATEGORICAL МЕТКАМИ")
print("=" * 80)

# Для классификации: кодируем raw таргет как класс
# Каждая уникальная пара (таргет_тип, значение) → класс
y_raw_values = np.array([d['target_raw'] for d in multi_task_data])

# Создаём составной класс: "organism_metric_value"
composite_labels = [
    f"{d['organism']}_{d['metric_type']}_{d['target_raw']:.1f}"
    for d in multi_task_data
]
composite_encoder = LabelEncoder()
y_composite_cls = composite_encoder.fit_transform(composite_labels)

print(f"Уникальных составных классов: {len(composite_encoder.classes_)}")

X_train_cls, X_test_cls, y_train_cls, y_test_cls = train_test_split(
    X_multi_cat_df, y_composite_cls, test_size=0.2, random_state=42
)

train_pool_cls = Pool(data=X_train_cls, label=y_train_cls, cat_features=['organism', 'metric_type'])

cb_multi_cls = CatBoostClassifier(
    iterations=500, learning_rate=0.05, depth=6,
    verbose=False, random_seed=42, auto_class_weights='Balanced'
)
cb_multi_cls.fit(train_pool_cls)
y_pred_multi_cls = cb_multi_cls.predict(X_test_cls).flatten().astype(int)

acc_multi = accuracy_score(y_test_cls, y_pred_multi_cls)
f1_macro_multi = f1_score(y_test_cls, y_pred_multi_cls, average='macro', zero_division=0)
f1_weighted_multi = f1_score(y_test_cls, y_pred_multi_cls, average='weighted', zero_division=0)

print(f"\nMulti-Task Classification (Categorical):")
print(f"  Accuracy={acc_multi:.4f}  F1_macro={f1_macro_multi:.4f}  F1_weighted={f1_weighted_multi:.4f}")

# %% [markdown]
# ---
# # 🧪 ЭКСПЕРИМЕНТ 5: MULTI-TASK с OpenAI Embeddings
# ---

# %%
print("\n" + "=" * 80)
print("ЭКСПЕРИМЕНТ 5: MULTI-TASK С OPENAI EMBEDDINGS")
print("=" * 80)

# Текстовые описания классов для эмбеддингов
class_descriptions = {
    'S. aureus ATCC 43300_MIC': (
        "Minimum Inhibitory Concentration against Staphylococcus aureus ATCC 43300, "
        "a methicillin-resistant Gram-positive bacterium. MIC measures the lowest "
        "concentration that inhibits visible bacterial growth."
    ),
    'S. aureus ATCC 43300_MBC': (
        "Minimum Bactericidal Concentration against Staphylococcus aureus ATCC 43300, "
        "a methicillin-resistant Gram-positive bacterium. MBC measures the lowest "
        "concentration that kills 99.9% of bacteria."
    ),
    'E. coli ATCC 25922_MIC': (
        "Minimum Inhibitory Concentration against Escherichia coli ATCC 25922, "
        "a Gram-negative reference strain. MIC measures the lowest concentration "
        "that inhibits visible bacterial growth."
    ),
    'E. coli ATCC 25922_MBC': (
        "Minimum Bactericidal Concentration against Escherichia coli ATCC 25922, "
        "a Gram-negative reference strain. MBC measures the lowest concentration "
        "that kills 99.9% of bacteria."
    ),
}

print("\nПолучение OpenAI эмбеддингов для меток классов...")
class_embeddings_dict = {}

for key, description in class_descriptions.items():
    print(f"  {key}...", end=" ")
    embedding = get_embedding(description)
    class_embeddings_dict[key] = np.array(embedding)
    print(f"dim={len(embedding)}")

openai_emb_dim = len(class_embeddings_dict[list(class_embeddings_dict.keys())[0]])
print(f"\n✓ OpenAI эмбеддинги получены ({openai_emb_dim}-мерные)")

# %%
# --- 5a: REGRESSION с OpenAI embeddings ---
print("\n--- 5a: Multi-Task REGRESSION + OpenAI Embeddings ---")

X_multi_openai = []
for d in multi_task_data:
    mol_emb = d['molecular_emb']
    key = f"{d['organism']}_{d['metric_type']}"
    class_emb = class_embeddings_dict[key]
    X_multi_openai.append(np.concatenate([mol_emb, class_emb]))

X_multi_openai = np.array(X_multi_openai)
print(f"Размерность: {X_multi_openai.shape} (128 mol + {openai_emb_dim} openai)")

X_train_oai, X_test_oai, y_train_oai, y_test_oai = train_test_split(
    X_multi_openai, y_values, test_size=0.2, random_state=42
)

cb_oai_reg = CatBoostRegressor(iterations=500, learning_rate=0.05, depth=6,
                                verbose=False, random_seed=42)
cb_oai_reg.fit(X_train_oai, y_train_oai)
y_pred_oai = cb_oai_reg.predict(X_test_oai)

mae_oai = mean_absolute_error(y_test_oai, y_pred_oai)
rmse_oai = np.sqrt(mean_squared_error(y_test_oai, y_pred_oai))
r2_oai = r2_score(y_test_oai, y_pred_oai)
pearson_oai, _ = pearsonr(y_test_oai, y_pred_oai)

print(f"  MAE={mae_oai:.4f}  RMSE={rmse_oai:.4f}  R²={r2_oai:.4f}  Pearson={pearson_oai:.4f}")

# %%
# --- 5b: CLASSIFICATION с OpenAI embeddings ---
print("\n--- 5b: Multi-Task CLASSIFICATION + OpenAI Embeddings ---")

X_train_oai_c, X_test_oai_c, y_train_oai_c, y_test_oai_c = train_test_split(
    X_multi_openai, y_composite_cls, test_size=0.2, random_state=42
)

cb_oai_cls = CatBoostClassifier(
    iterations=500, learning_rate=0.05, depth=6,
    verbose=False, random_seed=42, auto_class_weights='Balanced'
)
cb_oai_cls.fit(X_train_oai_c, y_train_oai_c)
y_pred_oai_cls = cb_oai_cls.predict(X_test_oai_c).flatten().astype(int)

acc_oai = accuracy_score(y_test_oai_c, y_pred_oai_cls)
f1_macro_oai = f1_score(y_test_oai_c, y_pred_oai_cls, average='macro', zero_division=0)
f1_weighted_oai = f1_score(y_test_oai_c, y_pred_oai_cls, average='weighted', zero_division=0)

print(f"  Accuracy={acc_oai:.4f}  F1_macro={f1_macro_oai:.4f}  F1_weighted={f1_weighted_oai:.4f}")

# %% [markdown]
# ---
# # 📊 СРАВНИТЕЛЬНЫЙ АНАЛИЗ
# ---

# %%
print("\n" + "=" * 80)
print("                    СРАВНИТЕЛЬНЫЙ АНАЛИЗ")
print("=" * 80)

# --- REGRESSION ---
baseline_avg_mae = baseline_reg_df['MAE'].mean()
baseline_avg_rmse = baseline_reg_df['RMSE'].mean()
baseline_avg_r2 = baseline_reg_df['R²'].mean()
baseline_avg_pr = baseline_reg_df['Pearson_r'].mean()

regression_comparison = pd.DataFrame([
    {'Approach': '1. Baseline (отдельные модели)',
     'MAE': baseline_avg_mae, 'RMSE': baseline_avg_rmse,
     'R²': baseline_avg_r2, 'Pearson_r': baseline_avg_pr},
    {'Approach': '2. Multi-Task (Categorical)',
     'MAE': mae_cat, 'RMSE': rmse_cat,
     'R²': r2_cat, 'Pearson_r': pearson_cat},
    {'Approach': '3. Multi-Task (OpenAI Emb)',
     'MAE': mae_oai, 'RMSE': rmse_oai,
     'R²': r2_oai, 'Pearson_r': pearson_oai},
])

print("\n📈 REGRESSION:")
print(regression_comparison.to_string(index=False))

# --- CLASSIFICATION ---
baseline_cls_avg_acc = baseline_cls_df['Accuracy'].mean()
baseline_cls_avg_f1m = baseline_cls_df['F1_macro'].mean()
baseline_cls_avg_f1w = baseline_cls_df['F1_weighted'].mean()

classification_comparison = pd.DataFrame([
    {'Approach': '1. Baseline (отдельные модели)',
     'Accuracy': baseline_cls_avg_acc, 'F1_macro': baseline_cls_avg_f1m,
     'F1_weighted': baseline_cls_avg_f1w},
    {'Approach': '2. Multi-Task (Categorical)',
     'Accuracy': acc_multi, 'F1_macro': f1_macro_multi,
     'F1_weighted': f1_weighted_multi},
    {'Approach': '3. Multi-Task (OpenAI Emb)',
     'Accuracy': acc_oai, 'F1_macro': f1_macro_oai,
     'F1_weighted': f1_weighted_oai},
])

print("\n📊 CLASSIFICATION:")
print(classification_comparison.to_string(index=False))

# --- Улучшения ---
print("\n" + "─" * 60)
print("УЛУЧШЕНИЯ ОТНОСИТЕЛЬНО BASELINE")
print("─" * 60)

print("\nRegression R²:")
for _, row in regression_comparison.iterrows():
    delta = row['R²'] - baseline_avg_r2
    pct = (row['R²'] / baseline_avg_r2 - 1) * 100 if baseline_avg_r2 != 0 else 0
    print(f"  {row['Approach']:40s}  R²={row['R²']:.4f}  Δ={delta:+.4f}  ({pct:+.1f}%)")

print("\nClassification Accuracy:")
for _, row in classification_comparison.iterrows():
    delta = row['Accuracy'] - baseline_cls_avg_acc
    pct = (row['Accuracy'] / baseline_cls_avg_acc - 1) * 100 if baseline_cls_avg_acc != 0 else 0
    print(f"  {row['Approach']:40s}  Acc={row['Accuracy']:.4f}  Δ={delta:+.4f}  ({pct:+.1f}%)")

# %% [markdown]
# ## 📊 Визуализация

# %%
# === 1. Regression comparison bar chart ===
fig, axes = plt.subplots(1, 4, figsize=(20, 5))
colors = ['steelblue', 'coral', 'lightgreen']

for idx, metric in enumerate(['MAE', 'RMSE', 'R²', 'Pearson_r']):
    ax = axes[idx]
    vals = regression_comparison[metric].values
    bars = ax.bar(range(len(vals)), vals, color=colors, alpha=0.8, edgecolor='black')
    ax.set_ylabel(metric, fontsize=12, fontweight='bold')
    ax.set_title(f'Regression: {metric}', fontsize=13, fontweight='bold')
    ax.set_xticks(range(len(vals)))
    ax.set_xticklabels(['Baseline', 'Categorical', 'OpenAI'], rotation=15)
    ax.grid(axis='y', alpha=0.3)
    for bar, val in zip(bars, vals):
        ax.text(bar.get_x() + bar.get_width()/2., bar.get_height(),
                f'{val:.3f}', ha='center', va='bottom', fontsize=10, fontweight='bold')

plt.tight_layout()
plt.savefig(f'{BASE_PATH}/regression_comparison.png', dpi=300, bbox_inches='tight')
plt.show()

# %%
# === 2. Classification comparison bar chart ===
fig, axes = plt.subplots(1, 3, figsize=(16, 5))

for idx, metric in enumerate(['Accuracy', 'F1_macro', 'F1_weighted']):
    ax = axes[idx]
    vals = classification_comparison[metric].values
    bars = ax.bar(range(len(vals)), vals, color=colors, alpha=0.8, edgecolor='black')
    ax.set_ylabel(metric, fontsize=12, fontweight='bold')
    ax.set_title(f'Classification: {metric}', fontsize=13, fontweight='bold')
    ax.set_xticks(range(len(vals)))
    ax.set_xticklabels(['Baseline', 'Categorical', 'OpenAI'], rotation=15)
    ax.grid(axis='y', alpha=0.3)
    for bar, val in zip(bars, vals):
        ax.text(bar.get_x() + bar.get_width()/2., bar.get_height(),
                f'{val:.3f}', ha='center', va='bottom', fontsize=10, fontweight='bold')

plt.tight_layout()
plt.savefig(f'{BASE_PATH}/classification_comparison.png', dpi=300, bbox_inches='tight')
plt.show()

# %%
# === 3. Baseline per-target detail ===
fig, axes = plt.subplots(1, 2, figsize=(18, 6))

# Regression
ax = axes[0]
x_pos = np.arange(len(baseline_reg_df))
w = 0.2
ax.bar(x_pos - 1.5*w, baseline_reg_df['MAE'], w, label='MAE', color='steelblue', alpha=0.8)
ax.bar(x_pos - 0.5*w, baseline_reg_df['RMSE'], w, label='RMSE', color='coral', alpha=0.8)
ax.bar(x_pos + 0.5*w, baseline_reg_df['R²'], w, label='R²', color='lightgreen', alpha=0.8)
ax.bar(x_pos + 1.5*w, baseline_reg_df['Pearson_r'], w, label='Pearson', color='gold', alpha=0.8)
ax.set_title('Baseline Regression per Target', fontsize=13, fontweight='bold')
ax.set_xticks(x_pos)
ax.set_xticklabels([f"{r['Organism'].split(' ')[0]}\n{r['Target']}"
                     for _, r in baseline_reg_df.iterrows()], fontsize=9)
ax.legend()
ax.grid(axis='y', alpha=0.3)

# Classification
ax = axes[1]
x_pos = np.arange(len(baseline_cls_df))
w = 0.25
ax.bar(x_pos - w, baseline_cls_df['Accuracy'], w, label='Accuracy', color='steelblue', alpha=0.8)
ax.bar(x_pos, baseline_cls_df['F1_macro'], w, label='F1 macro', color='coral', alpha=0.8)
ax.bar(x_pos + w, baseline_cls_df['F1_weighted'], w, label='F1 weighted', color='lightgreen', alpha=0.8)
ax.set_title('Baseline Classification per Target', fontsize=13, fontweight='bold')
ax.set_xticks(x_pos)
ax.set_xticklabels([f"{r['Organism'].split(' ')[0]}\n{r['Target']}"
                     for _, r in baseline_cls_df.iterrows()], fontsize=9)
ax.legend()
ax.grid(axis='y', alpha=0.3)

plt.tight_layout()
plt.savefig(f'{BASE_PATH}/baseline_per_target_detail.png', dpi=300, bbox_inches='tight')
plt.show()

# %%
# === 4. Scatter: Predicted vs True (Regression) ===
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

# Baseline (первый таргет)
ax = axes[0]
tc = target_columns[0]
vi = ~target[tc].isna()
Xv = molecular_embeddings[vi]; yv = np.log2(target[tc][vi].values)
Xtr, Xte, ytr, yte = train_test_split(Xv, yv, test_size=0.2, random_state=42)
_cb = CatBoostRegressor(iterations=500, learning_rate=0.05, depth=6, verbose=False, random_seed=42)
_cb.fit(Xtr, ytr); _yp = _cb.predict(Xte)
ax.scatter(yte, _yp, alpha=0.6, s=50, edgecolors='black', linewidths=0.5, color='steelblue')
ax.plot([yte.min(), yte.max()], [yte.min(), yte.max()], 'r--', lw=2)
ax.set_xlabel('True log₂(MIC/MBC)'); ax.set_ylabel('Predicted')
ax.set_title(f'Baseline\nR²={r2_score(yte, _yp):.3f}', fontweight='bold')
ax.grid(alpha=0.3)

# Categorical
ax = axes[1]
ax.scatter(y_test_cat, y_pred_cat, alpha=0.6, s=50, edgecolors='black', linewidths=0.5, color='coral')
ax.plot([y_test_cat.min(), y_test_cat.max()], [y_test_cat.min(), y_test_cat.max()], 'r--', lw=2)
ax.set_xlabel('True log₂(MIC/MBC)'); ax.set_ylabel('Predicted')
ax.set_title(f'Multi-Task (Categorical)\nR²={r2_cat:.3f}', fontweight='bold')
ax.grid(alpha=0.3)

# OpenAI
ax = axes[2]
ax.scatter(y_test_oai, y_pred_oai, alpha=0.6, s=50, edgecolors='black', linewidths=0.5, color='lightgreen')
ax.plot([y_test_oai.min(), y_test_oai.max()], [y_test_oai.min(), y_test_oai.max()], 'r--', lw=2)
ax.set_xlabel('True log₂(MIC/MBC)'); ax.set_ylabel('Predicted')
ax.set_title(f'Multi-Task (OpenAI Emb)\nR²={r2_oai:.3f}', fontweight='bold')
ax.grid(alpha=0.3)

plt.tight_layout()
plt.savefig(f'{BASE_PATH}/predictions_scatter.png', dpi=300, bbox_inches='tight')
plt.show()

# %% [markdown]
# ## 🎯 Анализ важности признаков (OpenAI подход)

# %%
print("\n" + "=" * 80)
print("АНАЛИЗ ВАЖНОСТИ ПРИЗНАКОВ (OpenAI Regression)")
print("=" * 80)

fi = cb_oai_reg.get_feature_importance()
mol_fi = fi[:128]
oai_fi = fi[128:]

print(f"\nМолекулярные (128 dim):  mean={mol_fi.mean():.4f}  max={mol_fi.max():.4f}")
print(f"OpenAI ({openai_emb_dim} dim):  mean={oai_fi.mean():.4f}  max={oai_fi.max():.4f}")
print(f"Отношение OpenAI/Molecular: {oai_fi.mean() / mol_fi.mean():.2f}x")

fig, axes = plt.subplots(1, 2, figsize=(16, 5))

axes[0].bar(range(len(mol_fi)), mol_fi, alpha=0.7, color='steelblue')
axes[0].set_xlabel('Feature index'); axes[0].set_ylabel('Importance')
axes[0].set_title('Molecular Embeddings Importance', fontweight='bold')
axes[0].grid(alpha=0.3)

axes[1].bar(range(len(oai_fi)), oai_fi, alpha=0.7, color='lightgreen')
axes[1].set_xlabel('Feature index'); axes[1].set_ylabel('Importance')
axes[1].set_title('OpenAI Embeddings Importance', fontweight='bold')
axes[1].grid(alpha=0.3)

plt.tight_layout()
plt.savefig(f'{BASE_PATH}/feature_importance_openai.png', dpi=300, bbox_inches='tight')
plt.show()

# %% [markdown]
# ## 📋 Сохранение результатов

# %%
regression_comparison.to_excel(f'{BASE_PATH}/regression_comparison.xlsx', index=False)
classification_comparison.to_excel(f'{BASE_PATH}/classification_comparison.xlsx', index=False)
baseline_reg_df.to_excel(f'{BASE_PATH}/baseline_regression_per_target.xlsx', index=False)
baseline_cls_df.to_excel(f'{BASE_PATH}/baseline_classification_per_target.xlsx', index=False)

print("✓ Сохранены:")
print("  - regression_comparison.xlsx")
print("  - classification_comparison.xlsx")
print("  - baseline_regression_per_target.xlsx")
print("  - baseline_classification_per_target.xlsx")

# %% [markdown]
# ## 📋 Финальный отчёт

# %%
best_reg_approach = regression_comparison.loc[regression_comparison['R²'].idxmax(), 'Approach']
best_cls_approach = classification_comparison.loc[classification_comparison['Accuracy'].idxmax(), 'Approach']

report = f"""
{'='*80}
                    MULTI-TASK LEARNING — ФИНАЛЬНЫЙ ОТЧЁТ
{'='*80}

ЗАДАЧА: Предсказание антибактериальной активности (MIC/MBC) для 2 организмов

ПОДХОДЫ:
  1. Baseline — отдельные модели для каждого из 4 таргетов
  2. Multi-Task Categorical — объединённый датасет, метка класса как cat feature
  3. Multi-Task OpenAI — объединённый датасет, OpenAI эмбеддинги меток классов

ДАННЫЕ:
  - Молекулярные эмбеддинги: 128-мерные (Borderline SMOTE encoder)
  - OpenAI эмбеддинги: {openai_emb_dim}-мерные (text-embedding-3-small)
  - Multi-task датасет: {len(multi_task_data)} образцов

{'─'*80}
REGRESSION RESULTS:
{'─'*80}
{regression_comparison.to_string(index=False)}

Baseline per target:
{baseline_reg_df.to_string(index=False)}

{'─'*80}
CLASSIFICATION RESULTS:
{'─'*80}
{classification_comparison.to_string(index=False)}

Baseline per target:
{baseline_cls_df.to_string(index=False)}

{'─'*80}
ЛУЧШИЕ ПОДХОДЫ:
  Regression:      {best_reg_approach}
  Classification:  {best_cls_approach}

ВАЖНОСТЬ ПРИЗНАКОВ (OpenAI regression):
  Molecular mean importance: {mol_fi.mean():.4f}
  OpenAI mean importance:    {oai_fi.mean():.4f}
  Ratio OpenAI/Molecular:    {oai_fi.mean() / mol_fi.mean():.2f}x
{'='*80}
"""

with open(f'{BASE_PATH}/MULTI_TASK_FULL_REPORT.txt', 'w', encoding='utf-8') as f:
    f.write(report)

print(report)

# %%
print("✨ ЭКСПЕРИМЕНТ ЗАВЕРШЁН! ✨")
print(f"\n📈 Лучший regression:      {best_reg_approach}")
print(f"📊 Лучший classification:  {best_cls_approach}")
```
