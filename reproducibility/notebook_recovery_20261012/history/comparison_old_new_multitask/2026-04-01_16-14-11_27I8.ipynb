```python
import pandas as pd
import numpy as np
from sklearn.model_selection import KFold
from sklearn.metrics import mean_squared_error, r2_score
from sklearn.preprocessing import StandardScaler
from catboost import CatBoostRegressor
from sklearn.neural_network import MLPRegressor
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings('ignore')

# ═══════════════════════════════════════════════════════════════
# 0. ДАННЫЕ
# ═══════════════════════════════════════════════════════════════

df_old = pd.read_excel('/Users/egorilin/Desktop/COMA/Data_molformer_conc.xlsx')

# --- Признаки ---
feature_cols = [
    'Molecular weight', 'LogP', 'mmff94_dipole', 'gasteiger_dipole', 'eem2015bm_dipole',
    'Molformer_157', 'Molformer_197', 'Molformer_668', 'Molformer_85',
    'Molformer_127', 'Molformer_732', 'Molformer_543', 'Molformer_729',
    'Molformer_337', 'Molformer_686'
]

# --- Таргеты (подставь реальные названия колонок из df_old) ---
# Пример: MIC и MBC для S. aureus и E. coli
target_columns = [
    'S. aureus ATCC 43300 Activity MIC, мг/л',
    'S. aureus ATCC 43300 Activity MBC, мг/л',
    'E. coli ATCC 25922 Activity MIC, мг/л',
    'E. coli ATCC 25922 Activity MBC, мг/л'
]

# Краткие имена для удобства
short_names = ['SA_MIC', 'SA_MBC', 'EC_MIC', 'EC_MBC']

# Если у тебя несколько таргетов — раскомментируй выше и добавь нужные
# Чем больше таргетов, тем заметнее эффект multi-task

X_all = df_old[feature_cols].values

# ═══════════════════════════════════════════════════════════════
# 1. BASELINE — каждый таргет ОТДЕЛЬНО (кросс-валидация)
# ═══════════════════════════════════════════════════════════════

print("=" * 80)
print("BASELINE: КАЖДЫЙ ТАРГЕТ ОТДЕЛЬНО (5-Fold CV)")
print("=" * 80)

N_SPLITS = 5
kf = KFold(n_splits=N_SPLITS, shuffle=True, random_state=42)

baseline_results = []

for col, sn in zip(target_columns, short_names):
    valid_mask = ~df_old[col].isna()
    X = X_all[valid_mask]
    y = np.log2(df_old[col][valid_mask].values)

    fold_metrics = {'catboost': {'r2': [], 'rmse': []},
                    'mlp':      {'r2': [], 'rmse': []}}

    for fold, (tr_idx, te_idx) in enumerate(kf.split(X)):
        Xtr, Xte = X[tr_idx], X[te_idx]
        ytr, yte = y[tr_idx], y[te_idx]

        # --- CatBoost ---
        cb = CatBoostRegressor(iterations=500, learning_rate=0.05, depth=6,
                               verbose=False, random_seed=42)
        cb.fit(Xtr, ytr)
        yp_cb = cb.predict(Xte)
        fold_metrics['catboost']['r2'].append(r2_score(yte, yp_cb))
        fold_metrics['catboost']['rmse'].append(np.sqrt(mean_squared_error(yte, yp_cb)))

        # --- MLP ---
        scaler = StandardScaler()
        Xtr_s = scaler.fit_transform(Xtr)
        Xte_s = scaler.transform(Xte)

        mlp = MLPRegressor(hidden_layer_sizes=(256, 128, 64), max_iter=1000,
                           early_stopping=True, validation_fraction=0.15,
                           learning_rate='adaptive', learning_rate_init=0.001,
                           random_state=42)
        mlp.fit(Xtr_s, ytr)
        yp_mlp = mlp.predict(Xte_s)
        fold_metrics['mlp']['r2'].append(r2_score(yte, yp_mlp))
        fold_metrics['mlp']['rmse'].append(np.sqrt(mean_squared_error(yte, yp_mlp)))

    for model_name in ['catboost', 'mlp']:
        r2_mean = np.mean(fold_metrics[model_name]['r2'])
        r2_std  = np.std(fold_metrics[model_name]['r2'])
        rmse_mean = np.mean(fold_metrics[model_name]['rmse'])
        rmse_std  = np.std(fold_metrics[model_name]['rmse'])

        baseline_results.append({
            'target': sn, 'model': model_name, 'N': len(y),
            'R2_mean': r2_mean, 'R2_std': r2_std,
            'RMSE_mean': rmse_mean, 'RMSE_std': rmse_std,
        })
        print(f"  {sn:8s} | {model_name:10s} | R²={r2_mean:.4f}±{r2_std:.4f}  RMSE={rmse_mean:.4f}±{rmse_std:.4f}")

baseline_df = pd.DataFrame(baseline_results)


# ═══════════════════════════════════════════════════════════════
# 2. MULTI-TASK — все таргеты вместе + OneHot метка задачи
# ═══════════════════════════════════════════════════════════════

print("\n" + "=" * 80)
print("MULTI-TASK: ОБЪЕДИНЁННЫЙ ДАТАСЕТ + ONEHOT (5-Fold CV)")
print("=" * 80)

n_tasks = len(target_columns)

# Собираем единый датасет
rows_X, rows_y, rows_task, rows_orig_idx = [], [], [], []

for task_idx, col in enumerate(target_columns):
    valid_mask = ~df_old[col].isna()
    X = X_all[valid_mask]
    y = np.log2(df_old[col][valid_mask].values)
    indices = np.where(valid_mask)[0]

    onehot = np.zeros(n_tasks)
    onehot[task_idx] = 1.0

    for i in range(len(X)):
        rows_X.append(np.concatenate([X[i], onehot]))
        rows_y.append(y[i])
        rows_task.append(task_idx)
        rows_orig_idx.append(indices[i])

X_mt = np.array(rows_X)          # (N_total, n_features + n_tasks)
y_mt = np.array(rows_y)
task_ids = np.array(rows_task)
orig_idx = np.array(rows_orig_idx)

print(f"Объединённый датасет: {X_mt.shape[0]} образцов, {X_mt.shape[1]} признаков")
print(f"  ({len(feature_cols)} mol + {n_tasks} onehot)")
for i, sn in enumerate(short_names):
    print(f"  {sn}: {(task_ids == i).sum()} образцов")

# --- Кросс-валидация по уникальным молекулам ---
# Сплитим по orig_idx чтобы одна молекула не попала в train и test одновременно
unique_molecules = np.unique(orig_idx)
kf_mt = KFold(n_splits=N_SPLITS, shuffle=True, random_state=42)

mt_fold_metrics = {
    model: {sn: {'r2': [], 'rmse': []} for sn in short_names}
    for model in ['catboost', 'mlp']
}

for fold, (mol_tr_idx, mol_te_idx) in enumerate(kf_mt.split(unique_molecules)):
    train_mols = set(unique_molecules[mol_tr_idx])
    test_mols  = set(unique_molecules[mol_te_idx])

    train_mask = np.array([oi in train_mols for oi in orig_idx])
    test_mask  = np.array([oi in test_mols for oi in orig_idx])

    Xtr, Xte = X_mt[train_mask], X_mt[test_mask]
    ytr, yte = y_mt[train_mask], y_mt[test_mask]
    tasks_te = task_ids[test_mask]

    # --- CatBoost ---
    cb = CatBoostRegressor(iterations=500, learning_rate=0.05, depth=6,
                           verbose=False, random_seed=42)
    cb.fit(Xtr, ytr)
    yp_cb = cb.predict(Xte)

    # --- MLP ---
    scaler = StandardScaler()
    Xtr_s = scaler.fit_transform(Xtr)
    Xte_s = scaler.transform(Xte)

    mlp = MLPRegressor(hidden_layer_sizes=(256, 128, 64), max_iter=1000,
                       early_stopping=True, validation_fraction=0.15,
                       learning_rate='adaptive', learning_rate_init=0.001,
                       random_state=42)
    mlp.fit(Xtr_s, ytr)
    yp_mlp = mlp.predict(Xte_s)

    # Per-task метрики
    for i, sn in enumerate(short_names):
        mask = tasks_te == i
        if mask.sum() < 2:
            continue
        yt_i = yte[mask]

        for model_name, yp in [('catboost', yp_cb), ('mlp', yp_mlp)]:
            yp_i = yp[mask]
            mt_fold_metrics[model_name][sn]['r2'].append(r2_score(yt_i, yp_i))
            mt_fold_metrics[model_name][sn]['rmse'].append(np.sqrt(mean_squared_error(yt_i, yp_i)))

mt_results = []
for model_name in ['catboost', 'mlp']:
    for sn in short_names:
        r2s = mt_fold_metrics[model_name][sn]['r2']
        rmses = mt_fold_metrics[model_name][sn]['rmse']
        if len(r2s) == 0:
            continue
        mt_results.append({
            'target': sn, 'model': model_name,
            'R2_mean': np.mean(r2s), 'R2_std': np.std(r2s),
            'RMSE_mean': np.mean(rmses), 'RMSE_std': np.std(rmses),
        })
        print(f"  {sn:8s} | {model_name:10s} | R²={np.mean(r2s):.4f}±{np.std(r2s):.4f}  RMSE={np.mean(rmses):.4f}±{np.std(rmses):.4f}")

mt_df = pd.DataFrame(mt_results)


# ═══════════════════════════════════════════════════════════════
# 3. СРАВНИТЕЛЬНАЯ ТАБЛИЦА
# ═══════════════════════════════════════════════════════════════

print("\n" + "=" * 80)
print("СРАВНЕНИЕ: BASELINE vs MULTI-TASK")
print("=" * 80)

for model_name in ['catboost', 'mlp']:
    print(f"\n{'─'*60}")
    print(f"  {model_name.upper()}")
    print(f"{'─'*60}")
    print(f"  {'Target':8s} | {'BL_R²':>10s} | {'MT_R²':>10s} | {'ΔR²':>8s} | {'BL_RMSE':>10s} | {'MT_RMSE':>10s} | {'ΔRMSE':>8s}")
    print(f"  {'-'*8}-+-{'-'*10}-+-{'-'*10}-+-{'-'*8}-+-{'-'*10}-+-{'-'*10}-+-{'-'*8}")

    for sn in short_names:
        bl = baseline_df[(baseline_df['target'] == sn) & (baseline_df['model'] == model_name)]
        mt = mt_df[(mt_df['target'] == sn) & (mt_df['model'] == model_name)]
        if bl.empty or mt.empty:
            continue
        bl_r2   = bl['R2_mean'].values[0]
        mt_r2   = mt['R2_mean'].values[0]
        bl_rmse = bl['RMSE_mean'].values[0]
        mt_rmse = mt['RMSE_mean'].values[0]
        dr2   = mt_r2 - bl_r2
        drmse = bl_rmse - mt_rmse  # положительное = MT лучше
        print(f"  {sn:8s} | {bl_r2:10.4f} | {mt_r2:10.4f} | {dr2:+8.4f} | {bl_rmse:10.4f} | {mt_rmse:10.4f} | {drmse:+8.4f}")


# ═══════════════════════════════════════════════════════════════
# 4. ВИЗУАЛИЗАЦИЯ
# ═══════════════════════════════════════════════════════════════

fig, axes = plt.subplots(2, 2, figsize=(16, 12))

for row, metric, label, higher_better in [(0, 'R2_mean', 'R²', True),
                                           (1, 'RMSE_mean', 'RMSE', False)]:
    for col_idx, model_name in enumerate(['catboost', 'mlp']):
        ax = axes[row, col_idx]
        x = np.arange(len(short_names))
        w = 0.35

        bl_vals = [baseline_df[(baseline_df['target'] == sn) & (baseline_df['model'] == model_name)][metric].values[0]
                   for sn in short_names]
        mt_vals = [mt_df[(mt_df['target'] == sn) & (mt_df['model'] == model_name)][metric].values[0]
                   if not mt_df[(mt_df['target'] == sn) & (mt_df['model'] == model_name)].empty else 0
                   for sn in short_names]

        bars1 = ax.bar(x - w/2, bl_vals, w, label='Baseline', color='steelblue', alpha=0.85)
        bars2 = ax.bar(x + w/2, mt_vals, w, label='Multi-Task', color='coral', alpha=0.85)

        ax.set_ylabel(label, fontsize=12, fontweight='bold')
        arrow = '↑' if higher_better else '↓'
        ax.set_title(f'{model_name.upper()}: {label} per Target ({arrow} лучше)',
                     fontsize=13, fontweight='bold')
        ax.set_xticks(x)
        ax.set_xticklabels(short_names)
        ax.legend()
        ax.grid(axis='y', alpha=0.3)

        for b in list(bars1) + list(bars2):
            ax.text(b.get_x() + b.get_width()/2, b.get_height(),
                    f'{b.get_height():.3f}', ha='center', va='bottom', fontsize=9)

plt.tight_layout()
plt.savefig('baseline_vs_multitask_comparison.png', dpi=300, bbox_inches='tight')
plt.show()
print("✓ Сохранено: baseline_vs_multitask_comparison.png")
```

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import KFold
from sklearn.metrics import mean_squared_error, r2_score
from catboost import CatBoostRegressor
import time

# --- Данные ---
df_old = pd.read_excel('/Users/egorilin/Desktop/COMA/Data_molformer_conc.xlsx')

feature_cols = [
    'Molecular weight', 'LogP', 'mmff94_dipole', 'gasteiger_dipole', 'eem2015bm_dipole',
    'Molformer_157', 'Molformer_197', 'Molformer_668', 'Molformer_85',
    'Molformer_127', 'Molformer_732', 'Molformer_543', 'Molformer_729',
    'Molformer_337', 'Molformer_686'
]

target_col = 'S. aureus ATCC 43300 Activity MIC, мг/л'

valid_mask = ~df_old[target_col].isna()
X = df_old.loc[valid_mask, feature_cols].values
y = np.log2(df_old.loc[valid_mask, target_col].values)

print(f"X: {X.shape}, y: {y.shape}")

# --- Один фолд CatBoost ---
kf = KFold(n_splits=5, shuffle=True, random_state=42)
tr_idx, te_idx = next(iter(kf.split(X)))

start = time.time()

cb = CatBoostRegressor(iterations=500, learning_rate=0.05, depth=6,
                       verbose=100, random_seed=42)
cb.fit(X[tr_idx], y[tr_idx], eval_set=(X[te_idx], y[te_idx]))

elapsed = time.time() - start

yp = cb.predict(X[te_idx])
r2 = r2_score(y[te_idx], yp)
rmse = np.sqrt(mean_squared_error(y[te_idx], yp))

print(f"\nВремя: {elapsed:.1f} сек")
print(f"R²: {r2:.4f}, RMSE: {rmse:.4f}")
print(f"\nОценка полного пайплайна (5 фолдов × 2 подхода × 1 модель):")
print(f"  ~{elapsed * 10:.0f} сек ({elapsed * 10 / 60:.1f} мин)")
```
