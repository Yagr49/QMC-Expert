```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error, r2_score
from sklearn.preprocessing import StandardScaler
from catboost import CatBoostRegressor
from sklearn.neural_network import MLPRegressor
import warnings
warnings.filterwarnings('ignore')

# ═══════════════════════════════════════════════════════════════
# 0. ДАННЫЕ
# ═══════════════════════════════════════════════════════════════

df_old = pd.read_excel('/Users/egorilin/Desktop/COMA/Data_molformer_conc.xlsx')

feature_cols = [
    'Molecular weight', 'LogP', 'mmff94_dipole', 'gasteiger_dipole', 'eem2015bm_dipole',
    'Molformer_157', 'Molformer_197', 'Molformer_668', 'Molformer_85',
    'Molformer_127', 'Molformer_732', 'Molformer_543', 'Molformer_729',
    'Molformer_337', 'Molformer_686'
]

target_columns = [
    'S. aureus ATCC 43300 Activity MIC, мг/л',
    'S. aureus ATCC 43300 Activity MBC, мг/л',
    'E. coli ATCC 25922 Activity MIC, мг/л',
    'E. coli ATCC 25922 Activity MBC, мг/л'
]
short_names = ['SA_MIC', 'SA_MBC', 'EC_MIC', 'EC_MBC']

X_all = df_old[feature_cols].values

# ═══════════════════════════════════════════════════════════════
# 1. ПОИСК ЛУЧШЕГО random_state ДЛЯ КАЖДОГО ТАРГЕТА
# ═══════════════════════════════════════════════════════════════

print("=" * 80)
print("ПОИСК ЛУЧШЕЙ РАЗБИВКИ (random_state 0..99)")
print("=" * 80)

SEEDS = range(100)
TEST_SIZE = 0.2

best_seeds = {}  # {target: best_seed}
all_search_results = []

for col, sn in zip(target_columns, short_names):
    valid_mask = ~df_old[col].isna()
    X = X_all[valid_mask]
    y = np.log2(df_old[col][valid_mask].values)

    print(f"\n{sn} (N={len(y)}):")

    best_r2 = -999
    best_seed = 0
    seed_results = []

    for seed in SEEDS:
        Xtr, Xte, ytr, yte = train_test_split(X, y, test_size=TEST_SIZE, random_state=seed)

        scaler = StandardScaler()
        Xtr_s = scaler.fit_transform(Xtr)
        Xte_s = scaler.transform(Xte)

        # CatBoost
        cb = CatBoostRegressor(iterations=500, learning_rate=0.05, depth=6,
                               verbose=False, random_seed=42)
        cb.fit(Xtr_s, ytr)
        yp_cb = cb.predict(Xte_s)
        r2_cb = r2_score(yte, yp_cb)
        rmse_cb = np.sqrt(mean_squared_error(yte, yp_cb))

        # MLP
        mlp = MLPRegressor(hidden_layer_sizes=(256, 128, 64), max_iter=1000,
                           early_stopping=True, validation_fraction=0.15,
                           learning_rate='adaptive', learning_rate_init=0.001,
                           random_state=42)
        mlp.fit(Xtr_s, ytr)
        yp_mlp = mlp.predict(Xte_s)
        r2_mlp = r2_score(yte, yp_mlp)
        rmse_mlp = np.sqrt(mean_squared_error(yte, yp_mlp))

        avg_r2 = (r2_cb + r2_mlp) / 2

        seed_results.append({
            'target': sn, 'seed': seed,
            'R2_cb': r2_cb, 'RMSE_cb': rmse_cb,
            'R2_mlp': r2_mlp, 'RMSE_mlp': rmse_mlp,
            'R2_avg': avg_r2
        })

        if avg_r2 > best_r2:
            best_r2 = avg_r2
            best_seed = seed

    best_seeds[sn] = best_seed
    all_search_results.extend(seed_results)

    # Топ-5 сидов
    sorted_results = sorted(seed_results, key=lambda x: x['R2_avg'], reverse=True)
    print(f"  Top-5 seeds (by avg R²):")
    for r in sorted_results[:5]:
        print(f"    seed={r['seed']:3d}  CB: R²={r['R2_cb']:.4f} RMSE={r['RMSE_cb']:.4f}  |  MLP: R²={r['R2_mlp']:.4f} RMSE={r['RMSE_mlp']:.4f}  |  avg={r['R2_avg']:.4f}")
    print(f"  >>> ЛУЧШИЙ seed={best_seed}, avg R²={best_r2:.4f}")

search_df = pd.DataFrame(all_search_results)

# ═══════════════════════════════════════════════════════════════
# 2. ЛУЧШИЙ ЕДИНЫЙ SEED (по среднему R² по всем таргетам)
# ═══════════════════════════════════════════════════════════════

print("\n" + "=" * 80)
print("ЛУЧШИЙ ЕДИНЫЙ SEED (средний R² по всем таргетам)")
print("=" * 80)

seed_avg = search_df.groupby('seed')['R2_avg'].mean().reset_index()
seed_avg = seed_avg.sort_values('R2_avg', ascending=False)

print("\nТоп-10 единых seeds:")
for _, row in seed_avg.head(10).iterrows():
    s = int(row['seed'])
    avg = row['R2_avg']
    details = search_df[search_df['seed'] == s]
    det_str = "  ".join([f"{r['target']}: CB={r['R2_cb']:.3f} MLP={r['R2_mlp']:.3f}"
                         for _, r in details.iterrows()])
    print(f"  seed={s:3d}  avg_R²={avg:.4f}  | {det_str}")

BEST_GLOBAL_SEED = int(seed_avg.iloc[0]['seed'])
print(f"\n>>> ЛУЧШИЙ ЕДИНЫЙ SEED = {BEST_GLOBAL_SEED}")

# ═══════════════════════════════════════════════════════════════
# 3. BASELINE с лучшим seed'ом
# ═══════════════════════════════════════════════════════════════

print("\n" + "=" * 80)
print(f"BASELINE: ЛУЧШИЙ SEED = {BEST_GLOBAL_SEED}")
print("=" * 80)

baseline_results = []

for col, sn in zip(target_columns, short_names):
    valid_mask = ~df_old[col].isna()
    X = X_all[valid_mask]
    y = np.log2(df_old[col][valid_mask].values)

    Xtr, Xte, ytr, yte = train_test_split(X, y, test_size=TEST_SIZE, random_state=BEST_GLOBAL_SEED)

    scaler = StandardScaler()
    Xtr_s = scaler.fit_transform(Xtr)
    Xte_s = scaler.transform(Xte)

    # CatBoost
    cb = CatBoostRegressor(iterations=500, learning_rate=0.05, depth=6,
                           verbose=False, random_seed=42)
    cb.fit(Xtr_s, ytr)
    yp_cb = cb.predict(Xte_s)

    # MLP
    mlp = MLPRegressor(hidden_layer_sizes=(256, 128, 64), max_iter=1000,
                       early_stopping=True, validation_fraction=0.15,
                       learning_rate='adaptive', learning_rate_init=0.001,
                       random_state=42)
    mlp.fit(Xtr_s, ytr)
    yp_mlp = mlp.predict(Xte_s)

    for model_name, yp in [('catboost', yp_cb), ('mlp', yp_mlp)]:
        r2 = r2_score(yte, yp)
        rmse = np.sqrt(mean_squared_error(yte, yp))
        baseline_results.append({
            'target': sn, 'model': model_name,
            'R2': r2, 'RMSE': rmse
        })
        print(f"  {sn:8s} | {model_name:10s} | R²={r2:.4f}  RMSE={rmse:.4f}")

baseline_df = pd.DataFrame(baseline_results)

# ═══════════════════════════════════════════════════════════════
# 4. MULTI-TASK с лучшим seed'ом
# ═══════════════════════════════════════════════════════════════

print("\n" + "=" * 80)
print(f"MULTI-TASK: ЛУЧШИЙ SEED = {BEST_GLOBAL_SEED}")
print("=" * 80)

n_tasks = len(target_columns)
n_feat = len(feature_cols)

rows_X, rows_y, rows_task, rows_orig_idx = [], [], [], []

for task_idx, col in enumerate(target_columns):
    valid_mask = ~df_old[col].isna()
    X = X_all[valid_mask]
    y = np.log2(df_old[col][valid_mask].values)
    indices = np.where(valid_mask)[0]

    onehot = np.zeros(n_tasks)
    onehot[task_idx] = 1.0

    for i in range(len(X)):
        rows_X.append(np.concatenate([X[i], onehot]))
        rows_y.append(y[i])
        rows_task.append(task_idx)
        rows_orig_idx.append(indices[i])

X_mt = np.array(rows_X)
y_mt = np.array(rows_y)
task_ids = np.array(rows_task)
orig_idx = np.array(rows_orig_idx)

# Сплит по молекулам с тем же seed'ом
unique_molecules = np.unique(orig_idx)
train_mols, test_mols = train_test_split(unique_molecules, test_size=TEST_SIZE,
                                          random_state=BEST_GLOBAL_SEED)
train_mols_set = set(train_mols)
test_mols_set = set(test_mols)

train_mask = np.array([oi in train_mols_set for oi in orig_idx])
test_mask  = np.array([oi in test_mols_set for oi in orig_idx])

Xtr, Xte = X_mt[train_mask], X_mt[test_mask]
ytr, yte = y_mt[train_mask], y_mt[test_mask]
tasks_te = task_ids[test_mask]

print(f"Train: {Xtr.shape[0]}, Test: {Xte.shape[0]}")

# Нормализация (только признаки, не OneHot)
scaler = StandardScaler()
Xtr_scaled = Xtr.copy()
Xte_scaled = Xte.copy()
Xtr_scaled[:, :n_feat] = scaler.fit_transform(Xtr[:, :n_feat])
Xte_scaled[:, :n_feat] = scaler.transform(Xte[:, :n_feat])

# CatBoost
cb = CatBoostRegressor(iterations=500, learning_rate=0.05, depth=6,
                       verbose=False, random_seed=42)
cb.fit(Xtr_scaled, ytr)
yp_cb = cb.predict(Xte_scaled)

# MLP
mlp = MLPRegressor(hidden_layer_sizes=(256, 128, 64), max_iter=1000,
                   early_stopping=True, validation_fraction=0.15,
                   learning_rate='adaptive', learning_rate_init=0.001,
                   random_state=42)
mlp.fit(Xtr_scaled, ytr)
yp_mlp = mlp.predict(Xte_scaled)

mt_results = []
for i, sn in enumerate(short_names):
    mask = tasks_te == i
    if mask.sum() < 2:
        continue
    yt_i = yte[mask]
    for model_name, yp in [('catboost', yp_cb), ('mlp', yp_mlp)]:
        yp_i = yp[mask]
        r2 = r2_score(yt_i, yp_i)
        rmse = np.sqrt(mean_squared_error(yt_i, yp_i))
        mt_results.append({'target': sn, 'model': model_name, 'R2': r2, 'RMSE': rmse})
        print(f"  {sn:8s} | {model_name:10s} | R²={r2:.4f}  RMSE={rmse:.4f}")

mt_df = pd.DataFrame(mt_results)

# ═══════════════════════════════════════════════════════════════
# 5. ИТОГОВОЕ СРАВНЕНИЕ
# ═══════════════════════════════════════════════════════════════

print("\n" + "=" * 80)
print(f"ИТОГ: BASELINE vs MULTI-TASK (seed={BEST_GLOBAL_SEED})")
print("=" * 80)

for model_name in ['catboost', 'mlp']:
    print(f"\n{'─'*70}")
    print(f"  {model_name.upper()}")
    print(f"{'─'*70}")
    print(f"  {'Target':8s} | {'BL_R²':>8s} | {'MT_R²':>8s} | {'ΔR²':>8s} | {'BL_RMSE':>9s} | {'MT_RMSE':>9s} | {'ΔRMSE':>8s}")
    print(f"  {'-'*8}-+-{'-'*8}-+-{'-'*8}-+-{'-'*8}-+-{'-'*9}-+-{'-'*9}-+-{'-'*8}")

    for sn in short_names:
        bl = baseline_df[(baseline_df['target'] == sn) & (baseline_df['model'] == model_name)]
        mt = mt_df[(mt_df['target'] == sn) & (mt_df['model'] == model_name)]
        if bl.empty or mt.empty:
            continue
        bl_r2   = bl['R2'].values[0]
        mt_r2   = mt['R2'].values[0]
        bl_rmse = bl['RMSE'].values[0]
        mt_rmse = mt['RMSE'].values[0]
        print(f"  {sn:8s} | {bl_r2:8.4f} | {mt_r2:8.4f} | {mt_r2-bl_r2:+8.4f} | {bl_rmse:9.4f} | {mt_rmse:9.4f} | {bl_rmse-mt_rmse:+8.4f}")
```

```python
"""
=============================================================================
ПОЛНЫЙ ПАЙПЛАЙН: ОБУЧЕНИЕ ЭНКОДЕРА + MULTI-TASK ЭКСПЕРИМЕНТЫ
=============================================================================
Часть 1: Обучение контрастивного энкодера (Borderline SMOTE, dim=256)
         на данных классификации (Good/Bad)
Часть 2: Эксперименты на biocides данных:
         A. Baseline Regression — каждый таргет отдельно
         B. Multi-Task Regression — categorical метки
         C. Multi-Task Regression — OpenAI text embeddings меток
         D. То же для Classification
         Всё на 5-fold CV
=============================================================================
"""

import numpy as np
import pandas as pd
import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import Dataset, DataLoader
from sklearn.model_selection import StratifiedKFold, KFold
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.metrics import (
    accuracy_score, f1_score, mean_absolute_error,
    mean_squared_error, r2_score, silhouette_score
)
from imblearn.over_sampling import BorderlineSMOTE
from catboost import CatBoostRegressor, CatBoostClassifier, Pool
from collections import Counter
from scipy.stats import pearsonr
from openai import OpenAI
import matplotlib.pyplot as plt
from matplotlib.patches import FancyBboxPatch
from matplotlib.path import Path as MPath
import matplotlib.patches as mpatches
import matplotlib as mpl
import pickle
import time
import warnings
warnings.filterwarnings('ignore')

device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print(f"Device: {device}")

BASE = '/Users/egorilin/Desktop/COMA'
RANDOM_STATE = 42

# =============================================================================
# OpenRouter API
# =============================================================================
OPENROUTER_API_KEY = "[REDACTED_API_KEY]"
client = OpenAI(api_key=OPENROUTER_API_KEY, base_url="https://openrouter.ai/api/v1")
EMBEDDING_MODEL = "openai/text-embedding-3-small"

def get_embedding(text):
    resp = client.embeddings.create(input=[text.replace("\n", " ")], model=EMBEDDING_MODEL)
    return np.array(resp.data[0].embedding)

try:
    _test = get_embedding("test")
    print(f"✓ OpenRouter API OK (dim={len(_test)})")
    OPENAI_DIM = len(_test)
except Exception as e:
    print(f"❌ OpenRouter error: {e}")
    raise

# =============================================================================
# МОДЕЛИ
# =============================================================================
class ContrastiveLoss(nn.Module):
    def __init__(self, temp=0.05, alpha=0.3):
        super().__init__()
        self.temp, self.alpha = temp, alpha
    def forward(self, features, labels):
        bs = features.shape[0]
        features = F.normalize(features, dim=1)
        sim = torch.matmul(features, features.T) / self.temp
        labels = labels.view(-1, 1)
        mp = torch.eq(labels, labels.T).float().to(features.device)
        mn = 1 - mp
        dm = torch.scatter(torch.ones_like(mp), 1,
                           torch.arange(bs).view(-1, 1).to(features.device), 0)
        mp, mn = mp * dm, mn * dm
        exp_s = torch.exp(sim) * dm
        lp = sim - torch.log(exp_s.sum(1, keepdim=True))
        pt = (mp * lp).sum(1) / mp.sum(1).clamp(min=1.0)
        nw = F.softmax(torch.exp(sim) * mn / self.temp, dim=1)
        hn = (nw * sim * mn).sum(1)
        return (-(self.temp / 0.07) * pt + self.alpha * hn).mean()

class Encoder(nn.Module):
    def __init__(self, input_dim, latent=256):
        super().__init__()
        self.enc = nn.Sequential(
            nn.Linear(input_dim, 512), nn.BatchNorm1d(512), nn.ReLU(), nn.Dropout(0.3),
            nn.Linear(512, 384), nn.BatchNorm1d(384), nn.ReLU(), nn.Dropout(0.3),
            nn.Linear(384, latent), nn.BatchNorm1d(latent), nn.ReLU(), nn.Dropout(0.3))
        self.proj = nn.Sequential(
            nn.Linear(latent, latent), nn.BatchNorm1d(latent), nn.ReLU(),
            nn.Dropout(0.15), nn.Linear(latent, latent), nn.BatchNorm1d(latent))
    def forward(self, x, rp=True):
        f = self.enc(x); return (f, self.proj(f)) if rp else f

class DS(Dataset):
    def __init__(self, X, y, aug=False):
        self.X = torch.FloatTensor(X if isinstance(X, np.ndarray) else X.values)
        self.y = torch.LongTensor(y.flatten() if isinstance(y, np.ndarray) else y.values.flatten())
        self.aug = aug
    def __len__(self): return len(self.X)
    def __getitem__(self, i):
        x = self.X[i].clone()
        if self.aug and torch.rand(1).item() > 0.5: x += torch.randn_like(x) * 0.02
        return x, self.y[i]

def extract_emb(encoder, X_np):
    encoder.eval()
    out = []
    with torch.no_grad():
        for b in DataLoader(torch.FloatTensor(X_np), batch_size=256):
            out.append(encoder(b.to(device), rp=False).cpu().numpy())
    return np.vstack(out)


# #############################################################################
# ЧАСТЬ 1: ОБУЧЕНИЕ ЭНКОДЕРА НА CLASSIFICATION ДАННЫХ
# #############################################################################
print("\n" + "=" * 70)
print("ЧАСТЬ 1: ОБУЧЕНИЕ КОНТРАСТИВНОГО ЭНКОДЕРА (dim=256)")
print("=" * 70)

# Загрузка classification данных
m1 = pd.read_excel(f'{BASE}/Test_molformer-2.xlsx').drop(['Unnamed: 0'], axis=1)
m2 = pd.read_excel(f'{BASE}/Test_molformer_3.xlsx').drop(['Unnamed: 0'], axis=1)
if list(m1.columns) != list(m2.columns):
    cc = list(set(m1.columns) & set(m2.columns)); m1, m2 = m1[cc], m2[cc]
X_cls = pd.concat([m1, m2], ignore_index=True).values

t1 = pd.read_excel(f'{BASE}/2025-12-15_QACs_SMILES_Fixed.xlsx')
t2 = pd.read_excel(f'{BASE}/2026-02-11_QACs SMILES_NF_MS_519-1103.xls')
y_cls = (pd.concat([t1['Annotation_Final'], t2['Annotation_Final']],
                    ignore_index=True) == 'Good').astype(int).values

print(f"Classification data: {X_cls.shape}, balance: {dict(Counter(y_cls))}")

# Обучение на ВСЕХ данных (для максимального качества энкодера)
scaler_cls = StandardScaler()
X_cls_s = scaler_cls.fit_transform(X_cls)

X_bal, y_bal = BorderlineSMOTE(
    random_state=RANDOM_STATE, sampling_strategy=0.8, k_neighbors=5
).fit_resample(X_cls_s, y_cls)
print(f"After B-SMOTE: {dict(Counter(y_bal))}")

print("Training encoder (80 epochs)...")
loader = DataLoader(DS(X_bal, y_bal, aug=True), batch_size=64, shuffle=True)
encoder = Encoder(X_cls_s.shape[1], latent=256).to(device)
loss_fn = ContrastiveLoss()
opt = torch.optim.AdamW(encoder.parameters(), lr=1e-3, weight_decay=1e-4)
sched = torch.optim.lr_scheduler.CosineAnnealingLR(opt, T_max=80, eta_min=1e-6)

best_loss, best_st, pat = float('inf'), None, 0
for ep in range(80):
    encoder.train(); el = 0
    for xb, yb in loader:
        xb, yb = xb.to(device), yb.to(device)
        opt.zero_grad(); _, p = encoder(xb)
        loss = loss_fn(p, yb); loss.backward()
        nn.utils.clip_grad_norm_(encoder.parameters(), 1.0); opt.step()
        el += loss.item()
    sched.step(); avg = el / len(loader)
    if avg < best_loss:
        best_loss, pat = avg, 0
        best_st = {k: v.clone() for k, v in encoder.state_dict().items()}
    else:
        pat += 1
        if pat >= 15: break

encoder.load_state_dict(best_st)
print(f"✓ Encoder trained (loss: {best_loss:.4f})")

# Silhouette на всех данных
emb_all = extract_emb(encoder, X_cls_s)
sil = silhouette_score(emb_all, y_cls)
print(f"  Silhouette: {sil:.4f}")

# Сохранение
torch.save(encoder.state_dict(), f'{BASE}/encoder_256_final.pth')
with open(f'{BASE}/scaler_cls_final.pkl', 'wb') as f:
    pickle.dump(scaler_cls, f)
print(f"✓ Saved: encoder_256_final.pth, scaler_cls_final.pkl")


# #############################################################################
# ЧАСТЬ 2: ЗАГРУЗКА BIOCIDES ДАННЫХ + ИЗВЛЕЧЕНИЕ ЭМБЕДДИНГОВ
# #############################################################################
print("\n" + "=" * 70)
print("ЧАСТЬ 2: BIOCIDES ДАННЫЕ")
print("=" * 70)

data_bio = pd.read_excel(f"{BASE}/Test_molformer_original_biocides.xlsx").drop(['Unnamed: 0'], axis=1)
target_bio = pd.read_excel(f"{BASE}/Data_biocides.xlsx")

print(f"Biocides: {data_bio.shape[0]} samples, {data_bio.shape[1]} features")

target_columns = [
    'S. aureus ATCC 43300 Activity MIC, мг/л',
    'S. aureus ATCC 43300 Activity MBC, мг/л',
    'E. coli ATCC 25922 Activity MIC, мг/л',
    'E. coli ATCC 25922 Activity MBC, мг/л'
]

for col in target_columns:
    vals = sorted(target_bio[col].dropna().unique())
    print(f"  {col}: {len(vals)} unique, NaN={target_bio[col].isna().sum()}")

# Извлечение молекулярных эмбеддингов через обученный энкодер
data_bio_scaled = scaler_cls.transform(data_bio)
mol_emb = extract_emb(encoder, data_bio_scaled)
print(f"\n✓ Molecular embeddings: {mol_emb.shape}")


# #############################################################################
# ЧАСТЬ 3: OpenAI ЭМБЕДДИНГИ МЕТОК
# #############################################################################
print("\n" + "=" * 70)
print("ЧАСТЬ 3: OpenAI ЭМБЕДДИНГИ МЕТОК")
print("=" * 70)

class_descriptions = {
    'S. aureus ATCC 43300_MIC': (
        "Minimum Inhibitory Concentration against Staphylococcus aureus ATCC 43300, "
        "a methicillin-resistant Gram-positive bacterium. MIC measures the lowest "
        "concentration that inhibits visible bacterial growth."
    ),
    'S. aureus ATCC 43300_MBC': (
        "Minimum Bactericidal Concentration against Staphylococcus aureus ATCC 43300, "
        "a methicillin-resistant Gram-positive bacterium. MBC measures the lowest "
        "concentration that kills 99.9% of bacteria."
    ),
    'E. coli ATCC 25922_MIC': (
        "Minimum Inhibitory Concentration against Escherichia coli ATCC 25922, "
        "a Gram-negative reference strain. MIC measures the lowest concentration "
        "that inhibits visible bacterial growth."
    ),
    'E. coli ATCC 25922_MBC': (
        "Minimum Bactericidal Concentration against Escherichia coli ATCC 25922, "
        "a Gram-negative reference strain. MBC measures the lowest concentration "
        "that kills 99.9% of bacteria."
    ),
}

print("Getting OpenAI embeddings for task labels...")
class_emb_dict = {}
for key, desc in class_descriptions.items():
    class_emb_dict[key] = get_embedding(desc)
    print(f"  ✓ {key} ({len(class_emb_dict[key])}d)")


# #############################################################################
# ЧАСТЬ 4: ПОСТРОЕНИЕ ДАТАСЕТОВ
# #############################################################################
print("\n" + "=" * 70)
print("ЧАСТЬ 4: ПОСТРОЕНИЕ ДАТАСЕТОВ")
print("=" * 70)

# Per-target данные (для baseline)
per_target = {}
for col in target_columns:
    valid = ~target_bio[col].isna()
    organism = col.split(' Activity ')[0]
    metric = col.split(' Activity ')[1].replace(', мг/л', '')
    short = f"{organism.split(' ')[0]}_{metric}"

    per_target[short] = {
        'X': mol_emb[valid],
        'y_reg': np.log2(target_bio[col][valid].values),
        'y_cls': LabelEncoder().fit_transform(target_bio[col][valid].values),
        'le': LabelEncoder().fit(target_bio[col][valid].values),
        'n': valid.sum(),
        'full_name': col,
        'organism': organism,
        'metric': metric,
    }
    print(f"  {short}: {per_target[short]['n']} samples, "
          f"classes={len(per_target[short]['le'].classes_)}")

# Multi-task датасет
mt_rows = []
for col in target_columns:
    valid = ~target_bio[col].isna()
    X_v = mol_emb[valid]
    y_v = np.log2(target_bio[col][valid].values)
    y_raw = target_bio[col][valid].values
    organism = col.split(' Activity ')[0]
    metric = col.split(' Activity ')[1].replace(', мг/л', '')
    key = f"{organism}_{metric}"

    for i in range(len(X_v)):
        mt_rows.append({
            'mol': X_v[i], 'y_reg': y_v[i], 'y_raw': y_raw[i],
            'organism': organism, 'metric': metric, 'key': key,
        })

print(f"\nMulti-task dataset: {len(mt_rows)} rows")

X_mt_mol = np.array([r['mol'] for r in mt_rows])
y_mt_reg = np.array([r['y_reg'] for r in mt_rows])
y_mt_raw = np.array([r['y_raw'] for r in mt_rows])

org_enc = LabelEncoder(); met_enc = LabelEncoder()
org_coded = org_enc.fit_transform([r['organism'] for r in mt_rows])
met_coded = met_enc.fit_transform([r['metric'] for r in mt_rows])

# Categorical version
X_mt_cat = pd.DataFrame(X_mt_mol, columns=[f'e{i}' for i in range(256)])
X_mt_cat['org'] = [str(x) for x in org_coded]
X_mt_cat['met'] = [str(x) for x in met_coded]

# OpenAI version
oai_features = np.array([class_emb_dict[r['key']] for r in mt_rows])
X_mt_oai = np.hstack([X_mt_mol, oai_features])

# Composite class labels
comp_labels = [f"{r['key']}_{r['y_raw']:.1f}" for r in mt_rows]
comp_enc = LabelEncoder()
y_mt_cls = comp_enc.fit_transform(comp_labels)

print(f"  Categorical features: {X_mt_cat.shape}")
print(f"  OpenAI features: {X_mt_oai.shape}")
print(f"  Composite classes: {len(comp_enc.classes_)}")


# #############################################################################
# ЧАСТЬ 5: ЭКСПЕРИМЕНТЫ НА 5-FOLD CV
# #############################################################################
N_FOLDS = 5

print("\n" + "=" * 70)
print(f"ЧАСТЬ 5: ЭКСПЕРИМЕНТЫ (5-fold CV)")
print("=" * 70)

# ─── A. BASELINE REGRESSION ─────────────────────────────────────────────
print("\n" + "─" * 60)
print("A. BASELINE REGRESSION — каждый таргет отдельно")
print("─" * 60)

baseline_reg = {}
for short, data in per_target.items():
    kf = KFold(n_splits=N_FOLDS, shuffle=True, random_state=RANDOM_STATE)
    fold_m = {'mae': [], 'rmse': [], 'r2': [], 'pearson': []}

    for tr_idx, val_idx in kf.split(data['X']):
        cb = CatBoostRegressor(iterations=500, learning_rate=0.05, depth=6,
                               verbose=False, random_seed=RANDOM_STATE)
        cb.fit(data['X'][tr_idx], data['y_reg'][tr_idx])
        yp = cb.predict(data['X'][val_idx])
        yt = data['y_reg'][val_idx]

        fold_m['mae'].append(mean_absolute_error(yt, yp))
        fold_m['rmse'].append(np.sqrt(mean_squared_error(yt, yp)))
        fold_m['r2'].append(r2_score(yt, yp))
        pr, _ = pearsonr(yt, yp) if len(yt) > 2 else (0, 1)
        fold_m['pearson'].append(pr)

    baseline_reg[short] = {m: (np.mean(v), np.std(v)) for m, v in fold_m.items()}
    print(f"  {short:20s} | R2={baseline_reg[short]['r2'][0]:.4f}±{baseline_reg[short]['r2'][1]:.4f} "
          f"MAE={baseline_reg[short]['mae'][0]:.4f}±{baseline_reg[short]['mae'][1]:.4f}")


# ─── B. MULTI-TASK REGRESSION (Categorical) ────────────────────────────
print("\n" + "─" * 60)
print("B. MULTI-TASK REGRESSION — categorical метки")
print("─" * 60)

kf = KFold(n_splits=N_FOLDS, shuffle=True, random_state=RANDOM_STATE)
mt_cat_reg = {'mae': [], 'rmse': [], 'r2': [], 'pearson': []}

for tr_idx, val_idx in kf.split(X_mt_cat):
    pool_tr = Pool(X_mt_cat.iloc[tr_idx], y_mt_reg[tr_idx], cat_features=['org', 'met'])
    cb = CatBoostRegressor(iterations=500, learning_rate=0.05, depth=6,
                           verbose=False, random_seed=RANDOM_STATE)
    cb.fit(pool_tr)
    yp = cb.predict(X_mt_cat.iloc[val_idx])
    yt = y_mt_reg[val_idx]
    mt_cat_reg['mae'].append(mean_absolute_error(yt, yp))
    mt_cat_reg['rmse'].append(np.sqrt(mean_squared_error(yt, yp)))
    mt_cat_reg['r2'].append(r2_score(yt, yp))
    pr, _ = pearsonr(yt, yp)
    mt_cat_reg['pearson'].append(pr)

mt_cat_reg_summary = {m: (np.mean(v), np.std(v)) for m, v in mt_cat_reg.items()}
print(f"  R2={mt_cat_reg_summary['r2'][0]:.4f}±{mt_cat_reg_summary['r2'][1]:.4f} "
      f"MAE={mt_cat_reg_summary['mae'][0]:.4f}±{mt_cat_reg_summary['mae'][1]:.4f}")


# ─── C. MULTI-TASK REGRESSION (OpenAI Embeddings) ──────────────────────
print("\n" + "─" * 60)
print("C. MULTI-TASK REGRESSION — OpenAI text embeddings")
print("─" * 60)

mt_oai_reg = {'mae': [], 'rmse': [], 'r2': [], 'pearson': []}
for tr_idx, val_idx in kf.split(X_mt_oai):
    cb = CatBoostRegressor(iterations=500, learning_rate=0.05, depth=6,
                           verbose=False, random_seed=RANDOM_STATE)
    cb.fit(X_mt_oai[tr_idx], y_mt_reg[tr_idx])
    yp = cb.predict(X_mt_oai[val_idx])
    yt = y_mt_reg[val_idx]
    mt_oai_reg['mae'].append(mean_absolute_error(yt, yp))
    mt_oai_reg['rmse'].append(np.sqrt(mean_squared_error(yt, yp)))
    mt_oai_reg['r2'].append(r2_score(yt, yp))
    pr, _ = pearsonr(yt, yp)
    mt_oai_reg['pearson'].append(pr)

mt_oai_reg_summary = {m: (np.mean(v), np.std(v)) for m, v in mt_oai_reg.items()}
print(f"  R2={mt_oai_reg_summary['r2'][0]:.4f}±{mt_oai_reg_summary['r2'][1]:.4f} "
      f"MAE={mt_oai_reg_summary['mae'][0]:.4f}±{mt_oai_reg_summary['mae'][1]:.4f}")


# ─── D. BASELINE CLASSIFICATION ────────────────────────────────────────
print("\n" + "─" * 60)
print("D. BASELINE CLASSIFICATION — каждый таргет отдельно")
print("─" * 60)

baseline_cls = {}
for short, data in per_target.items():
    n_cls = len(np.unique(data['y_cls']))
    min_count = min(Counter(data['y_cls']).values())
    use_strat = min_count >= N_FOLDS

    if use_strat:
        kf_c = StratifiedKFold(n_splits=N_FOLDS, shuffle=True, random_state=RANDOM_STATE)
        splits = kf_c.split(data['X'], data['y_cls'])
    else:
        kf_c = KFold(n_splits=N_FOLDS, shuffle=True, random_state=RANDOM_STATE)
        splits = kf_c.split(data['X'])

    fold_m = {'acc': [], 'f1_macro': [], 'f1_weighted': []}
    for tr_idx, val_idx in splits:
        cb = CatBoostClassifier(iterations=500, learning_rate=0.05, depth=6, verbose=False,
                                random_seed=RANDOM_STATE, auto_class_weights='Balanced')
        cb.fit(data['X'][tr_idx], data['y_cls'][tr_idx])
        yp = cb.predict(data['X'][val_idx]).flatten().astype(int)
        yt = data['y_cls'][val_idx]
        fold_m['acc'].append(accuracy_score(yt, yp))
        fold_m['f1_macro'].append(f1_score(yt, yp, average='macro', zero_division=0))
        fold_m['f1_weighted'].append(f1_score(yt, yp, average='weighted', zero_division=0))

    baseline_cls[short] = {m: (np.mean(v), np.std(v)) for m, v in fold_m.items()}
    print(f"  {short:20s} | Acc={baseline_cls[short]['acc'][0]:.4f}±{baseline_cls[short]['acc'][1]:.4f} "
          f"F1w={baseline_cls[short]['f1_weighted'][0]:.4f}±{baseline_cls[short]['f1_weighted'][1]:.4f}")


# ─── E. MULTI-TASK CLASSIFICATION (Categorical) ────────────────────────
print("\n" + "─" * 60)
print("E. MULTI-TASK CLASSIFICATION — categorical")
print("─" * 60)

mt_cat_cls = {'acc': [], 'f1_macro': [], 'f1_weighted': []}
for tr_idx, val_idx in KFold(N_FOLDS, shuffle=True, random_state=RANDOM_STATE).split(X_mt_cat):
    pool_tr = Pool(X_mt_cat.iloc[tr_idx], y_mt_cls[tr_idx], cat_features=['org', 'met'])
    cb = CatBoostClassifier(iterations=500, learning_rate=0.05, depth=6, verbose=False,
                            random_seed=RANDOM_STATE, auto_class_weights='Balanced')
    cb.fit(pool_tr)
    yp = cb.predict(X_mt_cat.iloc[val_idx]).flatten().astype(int)
    yt = y_mt_cls[val_idx]
    mt_cat_cls['acc'].append(accuracy_score(yt, yp))
    mt_cat_cls['f1_macro'].append(f1_score(yt, yp, average='macro', zero_division=0))
    mt_cat_cls['f1_weighted'].append(f1_score(yt, yp, average='weighted', zero_division=0))

mt_cat_cls_s = {m: (np.mean(v), np.std(v)) for m, v in mt_cat_cls.items()}
print(f"  Acc={mt_cat_cls_s['acc'][0]:.4f}±{mt_cat_cls_s['acc'][1]:.4f} "
      f"F1w={mt_cat_cls_s['f1_weighted'][0]:.4f}±{mt_cat_cls_s['f1_weighted'][1]:.4f}")


# ─── F. MULTI-TASK CLASSIFICATION (OpenAI) ─────────────────────────────
print("\n" + "─" * 60)
print("F. MULTI-TASK CLASSIFICATION — OpenAI embeddings")
print("─" * 60)

mt_oai_cls = {'acc': [], 'f1_macro': [], 'f1_weighted': []}
for tr_idx, val_idx in KFold(N_FOLDS, shuffle=True, random_state=RANDOM_STATE).split(X_mt_oai):
    cb = CatBoostClassifier(iterations=500, learning_rate=0.05, depth=6, verbose=False,
                            random_seed=RANDOM_STATE, auto_class_weights='Balanced')
    cb.fit(X_mt_oai[tr_idx], y_mt_cls[tr_idx])
    yp = cb.predict(X_mt_oai[val_idx]).flatten().astype(int)
    yt = y_mt_cls[val_idx]
    mt_oai_cls['acc'].append(accuracy_score(yt, yp))
    mt_oai_cls['f1_macro'].append(f1_score(yt, yp, average='macro', zero_division=0))
    mt_oai_cls['f1_weighted'].append(f1_score(yt, yp, average='weighted', zero_division=0))

mt_oai_cls_s = {m: (np.mean(v), np.std(v)) for m, v in mt_oai_cls.items()}
print(f"  Acc={mt_oai_cls_s['acc'][0]:.4f}±{mt_oai_cls_s['acc'][1]:.4f} "
      f"F1w={mt_oai_cls_s['f1_weighted'][0]:.4f}±{mt_oai_cls_s['f1_weighted'][1]:.4f}")


# #############################################################################
# ЧАСТЬ 6: СВОДНЫЕ ТАБЛИЦЫ
# #############################################################################
print("\n\n" + "=" * 70)
print("СВОДНЫЕ ТАБЛИЦЫ")
print("=" * 70)

# --- REGRESSION ---
print("\n📈 REGRESSION (5-fold CV, mean ± std):")
print(f"{'Approach':45s} {'R2':>16s} {'RMSE':>16s} {'MAE':>16s} {'Pearson':>16s}")
print("─" * 112)

# Baseline average
bl_r2 = np.mean([v['r2'][0] for v in baseline_reg.values()])
bl_r2_s = np.mean([v['r2'][1] for v in baseline_reg.values()])
bl_rmse = np.mean([v['rmse'][0] for v in baseline_reg.values()])
bl_rmse_s = np.mean([v['rmse'][1] for v in baseline_reg.values()])
bl_mae = np.mean([v['mae'][0] for v in baseline_reg.values()])
bl_mae_s = np.mean([v['mae'][1] for v in baseline_reg.values()])
bl_pr = np.mean([v['pearson'][0] for v in baseline_reg.values()])
bl_pr_s = np.mean([v['pearson'][1] for v in baseline_reg.values()])
print(f"{'Baseline (avg of 4 separate models)':45s} {bl_r2:.4f}±{bl_r2_s:.4f}  {bl_rmse:.4f}±{bl_rmse_s:.4f}  {bl_mae:.4f}±{bl_mae_s:.4f}  {bl_pr:.4f}±{bl_pr_s:.4f}")

for short, data in baseline_reg.items():
    print(f"  └─ {short:41s} {data['r2'][0]:.4f}±{data['r2'][1]:.4f}  "
          f"{data['rmse'][0]:.4f}±{data['rmse'][1]:.4f}  "
          f"{data['mae'][0]:.4f}±{data['mae'][1]:.4f}  {data['pearson'][0]:.4f}±{data['pearson'][1]:.4f}")

m = mt_cat_reg_summary
print(f"{'Multi-Task + Categorical':45s} {m['r2'][0]:.4f}±{m['r2'][1]:.4f}  {m['rmse'][0]:.4f}±{m['rmse'][1]:.4f}  {m['mae'][0]:.4f}±{m['mae'][1]:.4f}  {m['pearson'][0]:.4f}±{m['pearson'][1]:.4f}")

m = mt_oai_reg_summary
print(f"{'Multi-Task + OpenAI Embeddings':45s} {m['r2'][0]:.4f}±{m['r2'][1]:.4f}  {m['rmse'][0]:.4f}±{m['rmse'][1]:.4f}  {m['mae'][0]:.4f}±{m['mae'][1]:.4f}  {m['pearson'][0]:.4f}±{m['pearson'][1]:.4f}")

# --- CLASSIFICATION ---
print(f"\n📊 CLASSIFICATION (5-fold CV, mean ± std):")
print(f"{'Approach':45s} {'Accuracy':>16s} {'F1 weighted':>16s} {'F1 macro':>16s}")
print("─" * 95)

bl_acc = np.mean([v['acc'][0] for v in baseline_cls.values()])
bl_acc_s = np.mean([v['acc'][1] for v in baseline_cls.values()])
bl_f1w = np.mean([v['f1_weighted'][0] for v in baseline_cls.values()])
bl_f1w_s = np.mean([v['f1_weighted'][1] for v in baseline_cls.values()])
bl_f1m = np.mean([v['f1_macro'][0] for v in baseline_cls.values()])
bl_f1m_s = np.mean([v['f1_macro'][1] for v in baseline_cls.values()])
print(f"{'Baseline (avg of 4 separate models)':45s} {bl_acc:.4f}±{bl_acc_s:.4f}  {bl_f1w:.4f}±{bl_f1w_s:.4f}  {bl_f1m:.4f}±{bl_f1m_s:.4f}")

for short, data in baseline_cls.items():
    print(f"  └─ {short:41s} {data['acc'][0]:.4f}±{data['acc'][1]:.4f}  "
          f"{data['f1_weighted'][0]:.4f}±{data['f1_weighted'][1]:.4f}  {data['f1_macro'][0]:.4f}±{data['f1_macro'][1]:.4f}")

m = mt_cat_cls_s
print(f"{'Multi-Task + Categorical':45s} {m['acc'][0]:.4f}±{m['acc'][1]:.4f}  {m['f1_weighted'][0]:.4f}±{m['f1_weighted'][1]:.4f}  {m['f1_macro'][0]:.4f}±{m['f1_macro'][1]:.4f}")

m = mt_oai_cls_s
print(f"{'Multi-Task + OpenAI Embeddings':45s} {m['acc'][0]:.4f}±{m['acc'][1]:.4f}  {m['f1_weighted'][0]:.4f}±{m['f1_weighted'][1]:.4f}  {m['f1_macro'][0]:.4f}±{m['f1_macro'][1]:.4f}")


# #############################################################################
# ЧАСТЬ 7: СОХРАНЕНИЕ
# #############################################################################
# Собираем всё в Excel
reg_rows = [
    {'Approach': 'Baseline (avg)', 'R2_mean': bl_r2, 'R2_std': bl_r2_s,
     'RMSE_mean': bl_rmse, 'RMSE_std': bl_rmse_s,
     'MAE_mean': bl_mae, 'MAE_std': bl_mae_s, 'Pearson_mean': bl_pr, 'Pearson_std': bl_pr_s},
    {'Approach': 'Multi-Task Categorical', **{f'{m}_mean': mt_cat_reg_summary[m][0] for m in ['r2','rmse','mae','pearson']},
     **{f'{m}_std': mt_cat_reg_summary[m][1] for m in ['r2','rmse','mae','pearson']}},
    {'Approach': 'Multi-Task OpenAI', **{f'{m}_mean': mt_oai_reg_summary[m][0] for m in ['r2','rmse','mae','pearson']},
     **{f'{m}_std': mt_oai_reg_summary[m][1] for m in ['r2','rmse','mae','pearson']}},
]
pd.DataFrame(reg_rows).to_excel(f'{BASE}/multitask_regression_cv.xlsx', index=False)

cls_rows = [
    {'Approach': 'Baseline (avg)', 'Acc_mean': bl_acc, 'Acc_std': bl_acc_s,
     'F1w_mean': bl_f1w, 'F1w_std': bl_f1w_s, 'F1m_mean': bl_f1m, 'F1m_std': bl_f1m_s},
    {'Approach': 'Multi-Task Categorical', 'Acc_mean': mt_cat_cls_s['acc'][0], 'Acc_std': mt_cat_cls_s['acc'][1],
     'F1w_mean': mt_cat_cls_s['f1_weighted'][0], 'F1w_std': mt_cat_cls_s['f1_weighted'][1],
     'F1m_mean': mt_cat_cls_s['f1_macro'][0], 'F1m_std': mt_cat_cls_s['f1_macro'][1]},
    {'Approach': 'Multi-Task OpenAI', 'Acc_mean': mt_oai_cls_s['acc'][0], 'Acc_std': mt_oai_cls_s['acc'][1],
     'F1w_mean': mt_oai_cls_s['f1_weighted'][0], 'F1w_std': mt_oai_cls_s['f1_weighted'][1],
     'F1m_mean': mt_oai_cls_s['f1_macro'][0], 'F1m_std': mt_oai_cls_s['f1_macro'][1]},
]
pd.DataFrame(cls_rows).to_excel(f'{BASE}/multitask_classification_cv.xlsx', index=False)

print(f"\n✓ multitask_regression_cv.xlsx")
print(f"✓ multitask_classification_cv.xlsx")

print("\n" + "=" * 70)
print("DONE!")
print("=" * 70)
```
